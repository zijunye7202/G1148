📘 Temps i memòria: Complexitat

## 🧠 Complexitat empírica: mesura del temps d'execució

Podem mesurar experimentalment el temps que triga a executar-se un fragment
de codi. Aquest tipus de mesura depèn de factors com:

- el maquinari;
- la càrrega del sistema;
- la versió de Python;
- les dades utilitzades;
- altres processos que s'estiguin executant.

Per tant, **mesurar temps no és el mateix que calcular la complexitat
asimptòtica** d'un algorisme.

### `time.perf_counter()`

El mòdul `time` de Python proporciona `perf_counter()`, un rellotge d'alta
resolució adequat per mesurar intervals de temps curts.

```python
import time

inici = time.perf_counter()
funcio1()
temps = time.perf_counter() - inici

print(temps)
```

Una única mesura pot variar entre execucions. Per fer comparacions més
fiables convé repetir l'experiment diverses vegades.

### `%timeit` en Jupyter

En un notebook Jupyter amb IPython podem utilitzar la instrucció màgica
`%timeit`, que executa repetidament una instrucció i n'estima el temps
d'execució.

```python
def quadrat(x):
    return x * x

%timeit for x in range(100): quadrat(x)
```

Per mesurar tota una cel·la podem utilitzar `%%timeit` a la primera línia:

```python
%%timeit

llista = []
for i in range(1000):
    llista.append(i**2)
```

Els resultats exactes dependran de l'ordinador i de l'entorn d'execució.

````python
import time

inici: float = time.perf_counter()

quadrats: list[int] = []

for i in range(1000):
    quadrats.append(i**2)

temps_transcorregut: float = time.perf_counter() - inici

print(temps_transcorregut)
````

````python
# Observa el temps necessari per construir una llista gran.

inici: float = time.perf_counter()

llista: list[int] = []

for i in range(5_000_000):
    llista.append(i)

temps_transcorregut: float = time.perf_counter() - inici

print(temps_transcorregut)
````

````python
def quadrat(x: int) -> int:
    """
    Calcula el quadrat d'un nombre enter.

    Parameters
    ----------
    x : int
        Nombre que es vol elevar al quadrat.

    Returns
    -------
    resultat : int
        Quadrat del nombre.
    """
    return x * x


%timeit for x in range(100): quadrat(x)
````

# 📘 🧠 Complexitat calculada

## Anàlisi de la complexitat

L'anàlisi de complexitat estudia com creix el cost d'un algorisme quan
augmenta la mida de l'entrada.

Normalment representarem la mida de l'entrada amb `n`.

La notació **O gran**, `O(f(n))`, descriu una cota asimptòtica del creixement
del cost. En aquesta assignatura analitzarem habitualment el **pitjor cas**,
però també podem parlar del millor cas o del cas mitjà.

En l'anàlisi asimptòtica:

- ignorem constants multiplicatives;
- ignorem termes de creixement inferior;
- ens interessa el comportament quan `n` es fa gran.

Per exemple:

```text
3n + 7        → O(n)
5n² + 3n + 2  → O(n²)
n³ + n log n  → O(n³)
```

### Complexitat d'una operació

Moltes operacions elementals tenen cost constant:

```python
x = 3
y = x + 1
```

i les considerarem **O(1)** dins del model simplificat que utilitzarem.

Tanmateix, algunes operacions aparentment senzilles sobre col·leccions
depenen de la mida de la col·lecció.

Per exemple:

```python
llista.append("a")  # O(1) amortitzat
del llista[4]       # O(n)
```

En el segon cas, Python pot haver de desplaçar els elements posteriors.

### Bloc condicional

```python
if condicio:
    operacio1
else:
    operacio2
```

En una anàlisi de **pitjor cas**, la complexitat del bloc serà la de la
branca més costosa.

Per exemple, si:

```text
operacio1 → O(n)
operacio2 → O(1)
```

el bloc condicional és:

```text
O(n)
```

### Conjunt d'instruccions consecutives

Si tenim:

```python
operacio1
operacio2
operacio3
```

els costos se sumen.

Per exemple:

```text
O(n²) + O(n) + O(1) = O(n²)
```

Només conservem el terme que creix més ràpidament.

### Bucles

```python
for i in range(n):
    operacio
```

Si `operacio` és O(1) i s'executa `n` vegades:

```text
n · O(1) = O(n)
```

### Bucles imbricats

```python
for i in range(n):
    for j in range(n):
        operacio
```

El bucle intern s'executa `n` vegades per cadascuna de les `n`
iteracions del bucle extern:

```text
n · n = n²
```

Per tant:

```text
O(n²)
```

Si els dos bucles depenen de mides diferents:

```python
for i in range(n):
    for j in range(m):
        operacio
```

la complexitat és:

```text
O(nm)
```

> **Nota:** dir que una operació és O(1) és una simplificació del model
> de cost. Per exemple, els enters de Python poden tenir una mida arbitrària
> i les operacions aritmètiques sobre nombres molt grans també tenen un cost
> que depèn del nombre de dígits.

````python
# Exemple 1 de càlcul de complexitat


def funcio1(n: int) -> None:
    """
    Mostra els nombres enters entre 0 i n - 1.

    Parameters
    ----------
    n : int
        Nombre d'iteracions que s'han de realitzar.

    Returns
    -------
    None
        La funció mostra els valors per pantalla.
    """
    for i in range(n):  # Executem un cop aquest for
        print(i)        # Executem n vegades aquesta instrucció 
````

funcio1:  $O(n)$

````python
# Exemple 2 de càlcul de complexitat

def funcio2(n: int, m: int) -> None:
    """
    Mostra parelles d'índexs generades per dos bucles imbricats.

    Parameters
    ----------
    n : int
        Determina aproximadament n²/2 iteracions del bucle extern.
    m : int
        Determina m² iteracions del bucle intern.

    Returns
    -------
    None
        La funció mostra les parelles generades per pantalla.
    """
    for i in range(1, n * n, 2):
        # Aquest for s'executa de 1 fins a n * n amb passos de 2 en 2
        for j in range(m * m, 0, -1):
            # Aquest s'executa m * m vegades
            # Com que un depèn de l'altre, per a cada valor 'i' executem m*m operacions
            # i, per tant, el total serà la multiplicació dels dos valors
            print(f"({i},{j})", end=" ")
        print()

funcio2(4, 3)
````

El bucle extern fa aproximadament $n^2/2$ iteracions i l'intern $m^2$:

funcio2:
$$
O\left(\frac{n^2}{2}m^2\right)
=
O(n^2m^2)
$$

La constant $1/2$ desapareix en la notació asimptòtica.

````python
# Exemple 3 de càlcul de complexitat

def funcio3(llista: list[int], numero: int) -> list[bool]:
                                # suposem n = len(llista)

    """
    Indica quines posicions d'una llista coincideixen amb un nombre.

    Parameters
    ----------
    llista : list[int]
        Llista de nombres enters que es vol analitzar.
    numero : int
        Nombre que es vol localitzar.

    Returns
    -------
    resultat : list[bool]
        Llista de booleans. Cada posició és True si el valor corresponent
        de la llista original coincideix amb numero.
    """
    if numero in llista:                 # n operacions
        resultat: list[bool] = []        

        for valor in llista:            # n operacions 
            resultat.append(valor == numero) # 2 operacions unàries

        return resultat

    return [False] * len(llista)  # n operacions


llista: list[int] = [1, 2, 5, -6, 3, -8, 7, -9, 5, 8, 5]

funcio3(llista, 5)

   
llista: List[Any] = [1, 2, 5, -6, 3, -8, 7, -9, 5, 8, 5]
funcio3(llista, 5)
````

En el pitjor cas:

- `numero in llista` és O(n);
- el recorregut de la llista és O(n);
- crear `[False] * len(llista)` també és O(n).

Per tant:

$$
O(n) + O(n) = O(2n) = O(n)
$$

La complexitat de `funcio3` és **O(n)**.

````python
# Exemple 4 de càlcul de complexitat


def funcio4(n: int) -> None:
    """
    Insereix valors successivament en una llista i calcula una suma.

    Parameters
    ----------
    n : int
        Mida inicial de la llista.

    Returns
    -------
    None
        La funció mostra la llista resultant i la suma calculada.
    """
    llista: list[int] = [0] * n  # inicialitzar fa n passos
    suma: int = 0

    for i in range(1, len(llista) + 1):  
        # Aquest for executarà la part interna n vegades 
        llista.insert(i, i)
        # Insertar val n (veure collections.ipynb)
        suma += llista[i - 1] + llista[i]
         # 2 accessos i una suma = 3 operacions    

    print(llista, suma)
    # Imprimir una llista val n i imprimir un enter, 1


funcio4(10)
````


### Complexitat de `funcio4`

Inicialitzar una llista de mida `n` costa O(n).

El bucle s'executa `n` vegades. A cada iteració es fa un `insert()` en una
llista, operació que pot requerir desplaçar O(n) elements.

Per tant:

$$
O\Big(n) + n \cdot O\Big(n) + O\Big(n)
$$

és:

$$
O\Big(n^2)
$$

La complexitat dominant de `funcio4` és **O\Big(n²)**.

## Algorismes d'ordenació i cerca

A continuació es mostren alguns dels principals algorismes d'ordenació
i cerca que veurem a teoria, juntament amb la seva complexitat asimptòtica.

### Mergesort 

Mergesort divideix recursivament la llista en dues parts i després les
fusiona de manera ordenada.

La seva complexitat temporal és:

$$
O(n \log n)
$$

tant en el millor cas com en el cas mitjà i el pitjor cas.

La implementació següent necessita memòria auxiliar per construir les
llistes resultants.

````python
def merge(
    esquerra: list[int],
    dreta: list[int],
) -> list[int]:
    """
    Fusiona dues llistes ordenades en una nova llista ordenada.

    Parameters
    ----------
    esquerra : list[int]
        Primera llista ordenada.
    dreta : list[int]
        Segona llista ordenada.

    Returns
    -------
    resultat : list[int]
        Nova llista que conté tots els elements ordenats.
    """
    resultat: list[int] = []
    i: int = 0
    j: int = 0

    while i < len(esquerra) and j < len(dreta):
        if esquerra[i] <= dreta[j]:
            resultat.append(esquerra[i])
            i += 1
        else:
            resultat.append(dreta[j])
            j += 1

    resultat.extend(esquerra[i:])
    resultat.extend(dreta[j:])

    return resultat
````

````python
def mergesort(llista: list[int]) -> list[int]:
    """
    Ordena una llista d'enters mitjançant l'algorisme Mergesort.

    Parameters
    ----------
    llista : list[int]
        Llista de nombres enters que es vol ordenar.

    Returns
    -------
    resultat : list[int]
        Nova llista amb els elements ordenats.
    """
    if len(llista) < 2:
        return llista.copy()

    mig: int = len(llista) // 2

    esquerra: list[int] = mergesort(llista[:mig])
    dreta: list[int] = mergesort(llista[mig:])

    return merge(esquerra, dreta)
````

````python
mergesort([9, 8, 4, 2, 7, -4, 3])
````

### Quicksort

Quicksort selecciona un **pivot** i divideix els elements en grups situats
a l'esquerra i a la dreta del pivot.

La seva complexitat depèn de com quedin equilibrades les particions:

- **cas mitjà:** O(n log n)
- **millor cas:** O(n log n)
- **pitjor cas:** O(n²)

Per tant, no és correcte afirmar simplement que Quicksort és sempre
O(n log n).

La implementació següent utilitza una estratègia basada en la mediana
entre el primer, el central i el darrer element per escollir el pivot.

````python
def particio(
    llista: list[int],
    primer: int,
    ultim: int,
) -> int:
    """
    Particiona una llista al voltant d'un pivot.

    La funció selecciona el pivot mitjançant la mediana entre el primer,
    el central i el darrer element, i modifica la llista in situ.

    Parameters
    ----------
    llista : list[int]
        Llista que es vol particionar.
    primer : int
        Índex inicial de la partició.
    ultim : int
        Índex final de la partició.

    Returns
    -------
    posicio_pivot : int
        Posició final del pivot després de la partició.
    """
    mig: int = (primer + ultim) // 2

    if llista[primer] > llista[mig]:
        llista[primer], llista[mig] = llista[mig], llista[primer]

    if llista[primer] > llista[ultim]:
        llista[primer], llista[ultim] = llista[ultim], llista[primer]

    if llista[mig] > llista[ultim]:
        llista[mig], llista[ultim] = llista[ultim], llista[mig]

    llista[mig], llista[primer] = llista[primer], llista[mig]

    pivot: int = primer
    i: int = primer + 1
    j: int = ultim

    while True:
        while i <= ultim and llista[i] <= llista[pivot]:
            i += 1

        while j >= primer and llista[j] > llista[pivot]:
            j -= 1

        if i >= j:
            break

        llista[i], llista[j] = llista[j], llista[i]

    llista[j], llista[pivot] = llista[pivot], llista[j]

    return j
````

````python
def quick_sort(llista: list[int]) -> None:
    """
    Ordena una llista in situ mitjançant Quicksort.

    Parameters
    ----------
    llista : list[int]
        Llista de nombres enters que es vol ordenar.

    Returns
    -------
    None
        La funció modifica directament la llista rebuda.
    """
    _quick_sort_recursiu(llista, 0, len(llista) - 1)


def _quick_sort_recursiu(
    llista: list[int],
    primer: int,
    ultim: int,
) -> None:
    """
    Ordena recursivament una secció d'una llista.

    Parameters
    ----------
    llista : list[int]
        Llista que s'està ordenant.
    primer : int
        Índex inicial de la secció.
    ultim : int
        Índex final de la secció.

    Returns
    -------
    None
        La funció modifica la llista in situ.
    """
    if ultim > primer:
        pivot: int = particio(llista, primer, ultim)

        _quick_sort_recursiu(llista, primer, pivot - 1)
        _quick_sort_recursiu(llista, pivot + 1, ultim)
````

````python
llista: list[int] = [9, 8, 4, 2, 7, -4, 3]

quick_sort(llista)

print(llista)
````

### Cerca lineal, de complexitat O(n)

````python
def cerca_lineal(
    element: int,
    llista: list[int],
) -> int:
    """
    Cerca la primera aparició d'un element en una llista.

    Parameters
    ----------
    element : int
        Element que es vol cercar.
    llista : list[int]
        Llista de nombres enters.

    Returns
    -------
    posicio : int
        Índex de l'element o -1 si no es troba.
    """
    for index, valor in enumerate(llista):
        if valor == element:
            return index

    return -1
````

````python
cerca_lineal(10, [9, 8, 4, 2, 7, -4, 3])
````

````python
### Cerca binària

La cerca binària té complexitat:

$$
O(\log n)
$$

perquè a cada pas descarta aproximadament la meitat dels elements.

> **Precondició:** la llista ha d'estar ordenada.
````

````python
def cerca_binaria(
    element: int,
    llista: list[int],
) -> int:
    """
    Cerca un element en una llista ordenada mitjançant cerca binària.

    Parameters
    ----------
    element : int
        Element que es vol cercar.
    llista : list[int]
        Llista ordenada de nombres enters.

    Returns
    -------
    posicio : int
        Índex de l'element o -1 si no es troba.
    """
    return _cerca_binaria_recursiva(
        element,
        llista,
        0,
        len(llista) - 1,
    )


def _cerca_binaria_recursiva(
    element: int,
    llista: list[int],
    inici: int,
    final: int,
) -> int:
    """
    Cerca recursivament un element dins d'un interval de la llista.

    Parameters
    ----------
    element : int
        Element que es vol cercar.
    llista : list[int]
        Llista ordenada de nombres enters.
    inici : int
        Índex inicial de l'interval de cerca.
    final : int
        Índex final de l'interval de cerca.

    Returns
    -------
    posicio : int
        Índex de l'element o -1 si no es troba.
    """
    if inici > final:
        return -1

    mig: int = (inici + final) // 2
    valor: int = llista[mig]

    if valor == element:
        return mig

    if element < valor:
        return _cerca_binaria_recursiva(
            element,
            llista,
            inici,
            mig - 1,
        )

    return _cerca_binaria_recursiva(
        element,
        llista,
        mig + 1,
        final,
    )
````
