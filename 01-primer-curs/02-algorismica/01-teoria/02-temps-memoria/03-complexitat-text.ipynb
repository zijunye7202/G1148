### Comprovació que els caràcters d'una cadena són únics

Volem determinar si tots els caràcters d'una cadena són diferents.

La primera implementació manté una llista amb els caràcters que ja ha
trobat. Per cada nou caràcter, recorre aquesta llista per comprovar si
ja hi apareix.

```python
# n = nombre de caràcters del text
# m = nombre de caràcters diferents

def caracters_unics(text: str) -> bool:
    """
    Comprova si tots els caràcters d'una cadena són diferents.

    La funció manté una llista amb els caràcters que ja s'han
    examinat i comprova si cada nou caràcter hi apareix.

    Parameters
    ----------
    text : str
        Cadena de text que es vol comprovar.

    Returns
    -------
    resultat : bool
        True si tots els caràcters són únics; False si algun
        caràcter apareix més d'una vegada.
    """
    caracters_vistos: list[str] = []  # O(1)

    for caracter in text:  # Fins a n iteracions
        for caracter_vist in caracters_vistos:  # Fins a n iteracions
            if caracter == caracter_vist:  # O(1)
                return False  # O(1)

        caracters_vistos.append(caracter)  # O(1) amortitzat

    return True  # O(1)
```
