Escriu una funció que retorni el factorial d’un número entrat per paràmetre. No usis cap llibreria de
Python.

```python
def factorial_nombre(n: float) -> float:

    # donat n! = n * (n-1) ... 1

    # farem un procés iteratiu amb un bucle que multiplica per cada un dels nombres menors a n fins a 1

    factorial: float = n

    i: int = n - 1

    while i >= 1:
        
        factorial *= i
        i -= 1

    
    return factorial

factorial_nombre(5)
```
