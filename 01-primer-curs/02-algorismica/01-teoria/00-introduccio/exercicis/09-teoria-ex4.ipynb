Escriu una funció que demani l’edat dels membres de la teva família o de la teva colla i que finalment mostri la mitjana (𝜇) i la desviació estàndard (𝜎).

```python
def mitjana_desviacio() -> None:

    llista_edats: list[int] = []
    for membre in ["pare", "mare", "germà", "germana"]:

        edat: int = int(input(f"Introdueix l'edat del teu/va {membre}: "))

        llista_edats.append(edat)

    sumatori_edats: int = 0
    nombre_casos: int = len(llista_edats)
    
    for i in llista_edats:

        sumatori_edats += i

    mitjana: float = sumatori_edats/nombre_casos

    # la desviació el calcularem per parts:
    # 1. farem el sumatori de (edatX - mitjana)^2
    # 2. dividim el sumatori / nombre de casos
    # 3. farem arrel de la divisió

    sumatori_calcul: int = 0

    for i in llista_edats:

        sumatori_calcul += (i - mitjana)**2

    desviacio = (sumatori_calcul / nombre_casos)**0.5

    print(f"La mitjana de les edats és {mitjana} amb desviació {desviacio}") 
        
mitjana_desviacio()
       
```
