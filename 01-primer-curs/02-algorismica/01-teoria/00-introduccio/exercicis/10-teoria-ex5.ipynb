Refés l’exercici 3, usant diccionaris per relacionar cada vocal amb accent amb la vocal corresponent sense accent.

```python
def paraula_en_text(paraula: str, text: str) -> None:

    diccionari_vocals: dict[str, str] = {
        "à" : "a",
        "á" : "a",
        "è" : "e",
        "é" : "e",
        "ì" : "i",
        "í" : "i",
        "ò" : "o",
        "ó" : "o",
        "ù" : "u",
        "ú" : "u",
    }

    paraula_normal: str = paraula.lower()
    text_normal: str = text.lower()

    for accent, sense_accent in diccionari_vocals.items():

        paraula_normal = paraula_normal.replace(accent, sense_accent)
        text_normal = text_normal.replace(accent, sense_accent)

    if paraula_normal in text_normal:
        print(f"La paraula {paraula} és al text {text}.")

    else:
        print(f"La paraula {paraula} no és al text {text}.")
    

paraula_en_text("hola", "HOLA Bón diÀ")
```

el bucle for crea dues variables assignant les claus i valors del diccionari de forma iterativa, llavors per a cadascun de les parelles clau-valor a cada iteració es comprova si la paraula conté accent o no, en cas que estigui accentuat la vocal es fa replace() per la vocal no accentuada, el procés és el mateix amb les demés vocals seguint l'ordre del diccionari, aleshores si imprimeixo la parella al final del bucle em surten els últims vocals del diccionari.
