# 📘 Introducció a la programació 📘Funcions

 ## 🧠 Com escriure una funció
 
Una **funció** permet agrupar un conjunt d'instruccions sota un nom perquè
les puguem executar quan sigui necessari.

La definició d'una funció comença amb `def`, seguida del nom de la funció,
els parèntesis amb els possibles paràmetres i dos punts:

```python
def nom_funcio():
    instruccions
```

El cos de la funció s'escriu amb una indentació de **4 espais**.

Definir una funció no n'executa el cos. Per executar-la, l'hem de
**cridar** explícitament:

```python
nom_funcio()
```

Una funció potser no retorna res. Per exemple:

```python
def saludar() -> None:
    print("Hola")
```

O bé pot calcular i retornar un resultat mitjançant `return`. Per exemple:

```python
def sumar(a: int, b: int) -> int:
    return a + b
```

Si una funció arriba al final sense executar cap `return`, Python retorna
implícitament `None`.

````python
def funcio1() -> None:
    """
    Mostra el text "hola" per pantalla.

    Returns
    -------
    None
        La funció només mostra un missatge per pantalla.
    """
    print("hola")


def funcio2() -> None:
    """
    Mostra el text "a" per pantalla.

    Returns
    -------
    None
        La funció només mostra un missatge per pantalla.
    """
    print("a")
````

````python
funcio1()
funcio2()
funcio1()
````

## 🧠 Paràmetres 

Els **paràmetres** permeten que una funció treballi amb valors diferents
cada vegada que la cridem.

Els paràmetres s'indiquen entre els parèntesis de la definició:

```python
def mostrar_suma(a: int, b: int) -> None:
    print(a + b)
```

Quan cridem la funció, proporcionem els valors corresponents:

```python
mostrar_suma(3, 5)
mostrar_suma(10, 20)
```

Els valors que proporcionem en una crida s'anomenen **arguments**.

A continuació compararem una funció sense paràmetres amb una altra que
rep els nombres que ha de sumar.

````python
def suma() -> None:
    """
    Calcula i mostra la suma dels nombres 5 i 3.

    Returns
    -------
    None
        La funció mostra el resultat per pantalla.
    """
    var1: int = 5
    var2: int = 3
    suma_nombres: int = var1 + var2

    print(suma_nombres)
````

Amb aquesta funció només podem sumar els valors `5` i `3`, que són els que estan escrits directament dins del codi de la funció.

````python
def suma2(var1: int, var2: int) -> int:
    """
    Calcula i mostra la suma de dos nombres enters.

    Parameters
    ----------
    var1 : int
        Primer nombre que es vol sumar.
    var2 : int
        Segon nombre que es vol sumar.

    Returns
    -------
    int
        La funció retorna la suma de var1 i var2.
    """
    suma_nombres: int = var1 + var2
    return suma_nombres
````

````python
suma2(3, 5)
suma2(1, 2)
````

Amb aquesta versió podem sumar valors diferents en cada crida, perquè els nombres es proporcionen mitjançant els paràmetres `var1` i `var2`.

## ✍️ 1. Exercici 1: divisió

Fes una funció que rebi dos nombres com a paràmetres i en retorni la divisió.

Recorda que l'operador `/` de Python retorna un valor de tipus `float`.

Pensa també què hauria de passar si el divisor és `0`.

````python
def dividir(dividend: float, divisor: float) -> float:
    """
    Divideix un nombre entre un altre.

    Parameters
    ----------
    dividend : float
        Nombre que es vol dividir.
    divisor : float
        Nombre pel qual es divideix el dividend.

    Returns
    -------
    resultat : float
        Resultat de la divisió.
    """
    return dividend / divisor


dividir (4, 2)
````

## 🧠 Descomposició en funcions

A mesura que els programes es fan més complexos, és habitual que una única funció acabi realitzant massa tasques diferents.

Per facilitar:

- la **reutilització** del codi;
- la **llegibilitat**;
- les **proves**;
- la detecció d'errors;
- el **manteniment**;

és convenient dividir els problemes complexos en funcions més petites, cadascuna amb una responsabilitat clara.

Quan ens enfrontem a un problema podem:

1. identificar les diferents tasques que cal resoldre;
2. separar les tasques que tenen sentit per si mateixes;
3. implementar-les com a funcions;
4. combinar-les per resoldre el problema complet.

Una funció pot, per tant, cridar altres funcions.

Aquest procés de dividir un problema gran en problemes més petits és una de les idees fonamentals del disseny d'algorismes i programes.

## ✍️ 2. Exercici 2: descomposició en funcions

Escriu una acció `salutacio()` que rebi:

- el nom d'una persona;
- la seva edat.

L'acció ha de saludar la persona i indicar-li quants anys li falten per
arribar als 100.

Utilitza una funció auxiliar `falta_100()` que, donada una edat, retorni
quants anys falten per arribar a 100.

Per exemple:

```python
salutacio("Aina", 25)
```

hauria de mostrar:

```text
Hola Aina, et falten 75 anys per arribar a 100.
```

L'objectiu de l'exercici és que `salutacio()` **utilitzi** el resultat
retornat per `falta_100()`.

````python
def falta_100(edat: int) -> int:
    """
    Calcula quants anys falten per arribar als 100.

    Parameters
    ----------
    edat : int
        Edat actual de la persona.

    Returns
    -------
    anys : int
        Nombre d'anys que falten per arribar als 100.
    """
    return 100 - edat

falta_100(4)
````

````python
def salutacio(nom: str, edat: int) -> None:
    """
    Saluda una persona i indica quants anys li falten per arribar a 100.

    Parameters
    ----------
    nom : str
        Nom de la persona.
    edat : int
        Edat actual de la persona.

    Returns
    -------
    None
        La funció mostra el missatge per pantalla.
    """
    print ("Hola", nom, "et falten", falta_100(edat), "per arribar als 100.")


salutacio("Ignasi", 25)
````
