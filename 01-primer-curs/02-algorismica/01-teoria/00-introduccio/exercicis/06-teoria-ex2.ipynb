- Escriu una funció que vagi demanant un número a l’usuari fins a que aquest entri 100.
- Per a cada número que entri ha de mostrar un missatge a pantalla:
    - has entrat un número molt petit (>=0 i <50)
    - has entrat un número molt gran (>=50)
    - has entrat un número negatiu (<0).
- Quan l’usuari entri 100 ha de mostrar totes les desenes de 0 a 100, 100 inclòs, i ja no ha de
demanar cap més número

```python
def demanar_numero_fins_100() -> None:

    numero: int = 0
    
    while numero != 100:
    
        numero: int = int(input("Introdueix un número de 0 a 100 (100 per finalitzar): "))

        if numero >= 0 and numero < 50:
            print("has entrat un número molt petit")
                
        elif numero >= 50:
            print("has entrat un número molt gran")
                
        elif numero <0:
            print("has entrat un número negatiu")
                          
    if numero == 100:

        for i in range (10, 101, 10):
            print(i)
            
demanar_numero_fins_100()
    
```
