## 🧠 Python: crida de funcions

```python
# Definició de la funció “area_quadrat”
def area_quadrat(costat:int) -> int:
    return costat**2

area_quadrat(5)


def suma(a:int, b:int) -> int: # Definició d'una funció "sumar" amb paràmetres (informació que rep una funció per fer-la servir) a i b
    return a + b # Retorna el resultat d'operar per addició els paràmetres

def dif(a:int, b:int) -> int: # Definició d'una funció "diferència" amb paràmetres a i b (són distintes que les de la funció "suma")
    return a - b # Retorna el resultat d'operar per diferència els paràmetres

def sumdif(a:int, b:int) -> int: # Definició d'una funció "suma i diferència"
    s:int = suma(a, b) # Creació d'una variable entera que pren com a valor la funció "suma" amb els arguments a i b
    # arguments: informació que es passa a la funció cridada (per tal d'assignar els valors dels paràmetres d'aquella funció) 
    d:int = dif(a, b)
    return s, d # Retorn de les variables s i d


a:int = 3 # Assignació d'una variable a (enter)
b:int = 3
print(sumdif(a, b)) # Mostrar per pantalla el resultat de la crida de funció "suma i diferència"
```

## 🧠 Python: llistes

```python
l1: list[object] = []  # Crea una llista buida
l1.append("gat")    # Afegeix un element al final de la llista
l1.append(2)

print(l1)

print("reverse:", l1.reverse(), l1)
# .reverse() inverteix l'ordre dels elements de la llista original.
# Retorna None perquè modifica la mateixa llista i no retorna una nova llista.

l2: list[int] = [4, 2, 8, 9, 1, 2, 4, 7]

print("sort:", l2.sort(), l2)
# .sort() ordena els elements de la llista original de menor a major.
# Retorna None perquè modifica la mateixa llista.

print("count:", l2.count(4), l2.count(8), l2.count(0))
# .count(valor) retorna quantes vegades apareix un valor dins de la llista.

print("index:", l2.index(9)) # == len(l2) - 1
# .index(valor) retorna l'índex de la primera aparició del valor indicat.

l2.remove(8)
# .remove(valor) elimina la primera aparició del valor indicat.
# Retorna None.

print("Després del remove:", l2)

print("pop:", l2.pop(5))
# .pop(index) elimina i retorna l'element que es troba a l'índex indicat.
# Si no indiquem cap índex, elimina i retorna l'últim element.

print("Després del pop:", l2)

print("in:", 9 in l2)
# in comprova si un valor es troba dins de la llista i retorna True o False.
```

## 🧠 Python: referències

```python
a:int = 4
b:int = 5
c:int = 6
print(id(a), id(b), id(c))  
d:str = "banana"
e:str = "banana"
print(id(d),id(e), id(d) == id(e))
f:list[int] = [1, 2, 3]
g:list[int] = [1, 2, 3]
print(id(f),id(g), id(f) == id(g)) 

# id() és un identificador de les variables, que són objectes en espais en memòria
# donat l'exemple f,g: aquestes tenen els mateixos valors però els seus identificadors són diferents
# f == g però id(f) != id(g)
```

## 🧠 Python: còpia profunda de variables

```python
a:list[int] = [1, 2, 3]
b:list[int] = a # a i b tenen el mateix objecte llista
print(a,b,id(a),id(b)) # per tant coincideixen el identificadors
b[2] = 4
print(a) # mostren els mateixos valors 
print(b)

print()
a:list[int] = [1, 2, 3]
b:list[int] = a[:]  # b ja no apunta a "a", n'ha fet una còpia profunda ("clon")
print(a,b,id(a),id(b)) # els identificadors no coincideixen ja que no són el mateix objecte llista
b[2] = 4 # mostren diferents valors
print(a)
print(b)

```
