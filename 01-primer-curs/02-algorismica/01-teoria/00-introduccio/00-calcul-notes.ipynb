## 🧠 Càlcul nota portfoli laboratori

```python
def calcul_portfoli_sobre_10(
    prob_temps: float,
    nota_prob1: float,
    nota_prob2: float
) -> float:
    """
    Calcula la nota del portfoli sobre 10 punts.

    :param prob_temps: Quants problemes s'han lliurat a temps
    :param nota_prob1: Nota de la primera prova (sobre 10).
    :param nota_prob2: Nota de la segona prova (sobre 10).
    :return: Nota total del portfoli sobre 10 punts.
    """
    return (
        min(5.0, prob_temps * 0.5) +  # El màxim d'aquesta part seran 5 punts
        nota_prob1 * 0.25 +
        nota_prob2 * 0.25
    )
```

```python
nota_portfoli_sobre_10:float = calcul_portfoli_sobre_10(8.0, 10.0, 7.0)
print(nota_portfoli_sobre_10)
```

### 🧠 Calcul global laboratori

```python
def calcul_global_lab_sobre_3(
    prova1: float,
    prova2: float,
    prova3: float,
    nota_portfoli: float
) -> float:
    """
    Calcula la nota global del laboratori sobre 3 punts,
    ponderant tres proves i la nota del portfoli.

    :param prova1: Nota de la primera prova sobre 10.
    :param prova2: Nota de la segona prova sobre 10.
    :param prova3: Nota de la tercera prova sobre 10.
    :param nota_portfoli: Nota del portfoli sobre 10.
    :return: Nota global sobre 3 punts.
    """
    return (
        prova1 * 0.05 +
        prova2 * 0.075 +
        prova3 * 0.075 +
        nota_portfoli * 0.1
    )
```

```python
nota_lab_sobre_3 = calcul_global_lab_sobre_3(9.0,7.0,3.5,8.25)
print(nota_lab_sobre_3)
print("sobre 10 seria: ",  nota_lab_sobre_3 / 3 * 10)
```

### 🧠 Càlcul de la nota d'un examen

```python
def calcul_nota_examen_sobre_10(
    part_practica: float,
    part_teorica: float
    ) -> float:
    """
    Calcula la nota final de l'examen.

    :param part_practica: Nota de la part pràctica (sobre 5).
    :param part_teorica: Nota de la part teòrica (sobre 5).
    :return: Nota final de l'examen sobre 10 punts.
    """
    if part_practica > 2.5 and part_teorica > 2.5:
        return part_practica + part_teorica
    else:
        return min(part_practica, part_teorica)
```

```python
nota_parcial = calcul_nota_examen_sobre_10(2, 5)
print(nota_parcial)
```

```python
nota_parcial = calcul_nota_examen_sobre_10(3, 5)
print(nota_parcial)
```

### 🧠 Càlcul nota final teoria

```python
def calcul_global_teoria_sobre_7(
    parcial: float,
    final: float
    ) -> float:
    """
    Calcula la nota global de teoria sobre 7 punts,
    ponderant els dos examens de teoria.

    :param parcial: Nota del parcial (sobre 10).
    :param final: Nota del final (sobre 10).
    :return: Nota global sobre 7 punts.
    """
    return (
        parcial * 0.3 +
        final * 0.4
    )
```

```python
nota_final_teoria_sobre_7 = calcul_global_teoria_sobre_7(8.2, 4.8)
print(nota_final_teoria_sobre_7)
print("sobre 10 seria: ",  nota_final_teoria_sobre_7 / 7 * 10)
```

## 🧪 Calcul nota assignatura

```python
def calcul_global_assignatura(
    nota_lab: float,
    nota_teoria: float
    ) -> float:
    """
    Calcula la nota global de l'assignatura.

    :param nota_lab: Nota de laboratori (sobre 3).
    :param nota_teoria: Nota de teoria (sobre 7).
    :return: Nota final.
    """
    nota_lab_10 = nota_lab / 3 * 10
    nota_teoria_10 = nota_teoria / 7 * 10
    if nota_lab_10 >= 4.0 and nota_teoria_10 >= 4.0:
        return nota_lab + nota_teoria
    else:
        return min(4.0, nota_lab + nota_teoria)
```

```python
calcul_global_assignatura(1, 3.5)
```

```python
calcul_global_assignatura(3, 7)
```

```python
calcul_global_assignatura(3, 2)
```

## 🧪 Em puc presentar a reavaluació?

```python
def reavaluacio(
    nota_global: float
    ) -> bool:
    """
    Calcula si et pots presentar a reavaluació.

    :param nota_global: Nota d'avaluació continuada (sobre 10).
    :return: Et pots presentar o no
    """   
    return nota_global >= 3.5        
```
