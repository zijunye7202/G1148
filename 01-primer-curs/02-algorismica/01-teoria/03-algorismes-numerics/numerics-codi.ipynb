```python
import random

def fermat(num:int, test_count:int) -> bool:
    if num == 1:
        return False
    for x in range(test_count):
        val:int = random.randint(1, num-1)     
        if pow(val, num-1) % num != 1:      
            return False                   
    return True
```

```python
fermat(19073,100)
```

```python
def generar_primer(n:int)->int:
    trobat:bool = False
    valor_minim:int = 2**(n-1)
    valor_maxim:int = 2**n
    while not trobat:
        p:int = random.randint(valor_minim, valor_maxim)
        trobat = fermat(p, 10)
    return p

          
```

```python
import math

def divisio(x:int,y:int) -> list[int]:
    if x <= 0:
        return 0, 0
    if y == 1: 
        return x, 0
    q, r = divisio(math.floor(x / 2), y)
    q = 2 * q              #desfem la divisió per 2
    r = 2 * r              #desfem la divisió per 2 
    if x % 2 != 0:
        r += 1             #recuperem el que hem perdut amb el floor
    if r >= y:
        r = r - y          
        q = q + 1          #aquí és on anem augmentant el quocient
    return [q]+[r]
```

```python
def modexp(x,y,N):
    import math
    if y == 0:
        return 1    # cas base x^0 dona 1
    z = modexp(x, math.floor(y/2), N)  
                  # dividim la potència per 2 fins arribar al cas base  
    if y%2 == 0:
        return (z**2)%N
                # anem fent les potències de 2 mòdul N
    else:
        return (x*(z**2))%N
                # la y inicial no és una potència de 2, 
                # cal multiplicar per x, també mòdul N
```

```python
import math
import random

def trobar_e_d(m:int)->list[int]:
    # e és un coprimer de m, és a dir si e és primer, m no és multiple de e
    # de és l'invers modular de e (d*e mod M és congruent amb 1)
    # això fa que elevar a e i després elevar a d recuperi l'M original
    candidats:list[int] = sedas_eratostentes(1200)
    trobat:bool = False
    while (not trobat):
        e:int  = random.choice(candidats)
        if math.gcd(e, m) == 1:  # podriem comprovar només que m % e != 0
            d:int = pow(e, -1, m)
            if d > 1:
                trobat = True
    return [e,d]
```

```python
def generar_claus()->list[int]:
    p:int = generar_primer(5)
    q:int = generar_primer(5)
    n:int = p * q
    m:int = (p-1)*(q-1)
    l:list[int] = trobar_e_d(m)
    e = l[0]
    d = l[1]
    return [[n,e],[n,d]]
```

```python
def encriptar_missatge(clau_publica:list[int],missatge:int)->int:
    if missatge < 0 or missatge >= clau_publica[0]:
        print("El missatge ha de tenir un valor entre 0 i n-1")
        return None
    return pow(missatge,clau_publica[1],clau_publica[0])
```
