 ## 🧠 Estructura condicional

En Python utilitzem `if` per executar un bloc de codi només quan es compleix
una determinada condició.

Podem afegir:

- `elif` per comprovar altres condicions si les anteriors són falses;
- `else` per indicar què s'ha de fer si cap de les condicions anteriors
  és certa.

L'estructura general és:

```python
if condicio1:
    instruccions
elif condicio2:
    instruccions
else:
    instruccions
```

Els blocs es defineixen mitjançant la **indentació**.

````python
# Prova de canviar el valor assignat a la variable c1
c1: int = 25

if c1 > 24:
    print(f"{c1} és més gran que 24")
````

````python
# Prova de canviar el valor assignat a la variable c1
c1: int = 25

if c1 > 24:
    print(f"{c1} és més gran que 24")
else:
    print(f"{c1} no és més gran que 24")
````

````python
# Prova de canviar el valor assignat a la variable c1
c1: int = 25

if c1 > 24:
    print(f"{c1} és més gran que 24")
else:
    if c1 > 15:
        print(f"{c1} és més gran que 15 i menor o igual que 24")
    else:
        print(f"{c1} és menor o igual que 15")
````

````python
# elif equival a combinar conceptualment un else i un nou if
c1: int = 25

if c1 > 24:
    print(f"{c1} és més gran que 24")
elif c1 > 15:
    print(f"{c1} és més gran que 15 i menor o igual que 24")
else:
    print(f"{c1} és menor o igual que 15")
````

## ✍️ 1. Exercici 1: qualificació d'una prova

Defineix una variable `qualificacio` amb un valor entre `0` i `10`.

Escriu una estructura condicional que mostri:

- si la qualificació és menor que `5`:  
  `"Has tret un XX, has suspès."`
- de `5` a menys de `7`:  
  `"Has tret un XX, has aprovat."`
- de `7` a menys de `9`:  
  `"Has tret un XX, has tret un notable."`
- de `9` a `10`:  
  `"Has tret un XX, has tret un excel·lent."`

Per a aquest exercici pots assumir que la qualificació sempre està entre
`0` i `10`.

````python
def qualificacio_prova() -> None:

    qualificacio: int = 4

    if qualificacio < 5:
        
        print(f" Has tret un {qualificacio}, has suspès.")

    elif qualificacio >= 5 and qualificacio < 7:

        print(f"Has tret un {qualificacio}, has aprovat.")

    elif qualificacio >= 7 and qualificacio < 9:

        print(f"Has tret un {qualificacio}, has tret un notable.")

    elif qualificacio >= 9 and qualificacio <= 10:

        print(f"Has tret un {qualificacio}, has tret un excel·lent.")

qualificacio_prova()
````

 ## 🧠 Estructures iteratives

Les estructures iteratives permeten repetir un bloc d'instruccions.

En Python utilitzarem principalment:

- `for`: per recórrer els elements d'un iterable;
- `while`: per repetir un bloc mentre es compleixi una condició.

 ## 🧠 Iteracions for

````python
# En el cas del for, definim prèviament la seqüència de la iteració:
for i in range(5):
    print(f"El valor de i és {i}")
````

````python
# En el range podem definir el valor d'inici, el valor de parada i el salt

for i in range(1,10,2):
    print(f"El valor de i és {i}")
````

````python
# La seqüència no té per què ser numèrica. 
# Per exemple, si volem imprimir totes les lletres diferents d'"a":

for c in 'Hola Món':
    if c != "a":
        print(f"{c}")
````

````python
# També podem recorrer una llista

for animal in ["gat","gos","canari"]:
    print(f"Una mascota habitual és el {animal}")
````

 ## 🧠 Iteracions While

````python
# El while es repeteix mentre la seva condició sigui certa.
i: int = 0

while i < 5:
    print(f"{i} és menor que 5")
    i += 1  # Modifiquem i perquè la iteració pugui acabar.
````

````python
# La condició del while pot combinar diverses expressions lògiques.
# La iteració s'atura quan la condició deixa de ser certa.
i: int = 0

while i < 5 and i != 3:
    print(f"{i} és menor que 5 i diferent de 3")
    i += 1
````

 ## 🧠 Definició de funcions

En Python definim una funció amb la paraula reservada `def`.

Per exemple:

```python
def saludar() -> None:
    print("Hola")
```

Una funció pot rebre **paràmetres**:

```python
def saludar(nom: str) -> None:
    print(f"Hola {nom}")
```

i pot retornar un resultat mitjançant `return`:

```python
def sumar(a: int, b: int) -> int:
    return a + b
```

Si una funció arriba al final sense executar cap `return`, Python retorna
implícitament `None`.

````python
# Definim una acció escriu_hola_mon que mostra el missatge "Hola Món".
def escriu_hola_mon() -> None:
    """
    Mostra el missatge "Hola Món" per pantalla.

    Returns
    -------
    None
        La funció només mostra un missatge.
    """
    print("Hola Món")



# Perquè l'acció s'executi, l'hem d'invocar.
escriu_hola_mon()
````

````python
# Definim una funció suma_deu que retorna el valor rebut més 10.
def suma_deu(valor: int) -> int:
    """
    Suma 10 a un nombre enter.

    Parameters
    ----------
    valor : int
        Nombre al qual es vol sumar 10.

    Returns
    -------
    resultat : int
        Valor original incrementat en 10.
    """
    return valor + 10


# Com que la funció retorna un valor, podem assignar-ne el resultat.
nou_numero: int = suma_deu(1)
print(nou_numero)
````

Les variables i funcions definides en cel·les que ja hem executat continuen
disponibles en les cel·les posteriors perquè el **kernel manté l'estat de
la sessió**.

a definir.

Cal recordar, però, que l'ordre d'execució de les cel·les és important.
Si reiniciem el kernel, haurem de tornar a executar les cel·les que
