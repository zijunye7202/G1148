## Introducció

Les **col·leccions** són estructures de dades que permeten agrupar diversos
valors en un únic objecte.

En aquesta assignatura treballarem principalment amb:

- **llistes** (`list`);
- **tuples** (`tuple`);
- **diccionaris** (`dict`).

Més endavant també podem trobar altres tipus de col·leccions, com els
**conjunts** (`set`).

````python
# Exemple
llista: list[object] = ["gat", 8, True, 0b000]
print(llista)

diccionari: dict[int, str] = {
    1: "hola",
    2: "adeu",
}
print(diccionari[2])
````

Les llistes són **seqüències** ordenades d'elements. Una llista pot contenir elements de diferents tipus (no és **homogènia**) i podem modificar, afegir o eliminar els seus elements (és **mutable**).

Les tuples són molt semblants a les llistes però són **immutables** (és a dir no es poden canviar de manera directa). Aquesta característica imposa certes restriccions en la programació però fa que el seu processament sigui molt més eficient i en molts casos redefinirem una llista com a **`tupla`** per guanyar eficiència. 

Els **`diccionaris`** són parelles de clau-valor que permeten recuperar un valor a partir de la seva clau de manera molt eficient. El **`diccionari`**, a l'igual que la llista es pot recorrer seqüencialment, però no té un ordre prefixat.

Tant les llistes com els **`diccionaris`** actuen com a rang de valors i es poden recorrer en una iteració.

## Operacions amb col·leccions

Cada tipus de col·lecció proporciona operacions per:

- crear-la;
- consultar-ne els elements;
- afegir o eliminar elements, si és mutable;
- recórrer-ne els elements;
- obtenir-ne la longitud;
- comprovar si conté un determinat element.

Les operacions disponibles depenen del tipus de col·lecció.

### Còpia de llistes

**Atenció:** una assignació entre variables **no copia la llista**.

Si fem:

```python
llista2 = llista1
```

les dues variables fan referència al **mateix objecte**. Per tant, modificar
la llista a través d'una variable també afectarà el que observem a través
de l'altra.

Podem crear una nova llista mitjançant:

```python
llista3 = llista1.copy()
```

o mitjançant *slicing*:

```python
llista3 = llista1[:]
```

En aquests dos casos es crea una nova llista.

Aquestes operacions fan una **còpia superficial** (*shallow copy*).
Si la llista conté altres objectes mutables, com ara llistes niades,
aquests objectes interns continuen compartint-se.

Exemple:

```python
llista1 = [1, 2, 3]
llista2 = llista1
llista3 = llista1.copy()

llista1.append(4)
```

Després de l'operació:

```text
llista1 -> [1, 2, 3, 4]
llista2 -> [1, 2, 3, 4]
llista3 -> [1, 2, 3]
```

Observa bé el codi anterior i digues perquè llista2 és diferent de llista3.

````python
# Assignació: les dues variables fan referència a la mateixa llista.
a: list[int] = [1, 2, 3]
print("a original:", a)

b: list[int] = a

a.append(4)

print("a modificada:", a)
print("b comparteix la mateixa llista:", b)

print("*****")

# Còpia superficial: es crea una nova llista.
a = [1, 2, 3]
print("a original:", a)

b = a.copy()

a.append(4)

print("a modificada:", a)
print("b és una còpia independent:", b)
````

#### Slicing: o partir la llista en trossos

````python
# Podem triar una part de la llista indicant un rang.
#
# start: primer índex que s'inclou
# stop: índex final, que no s'inclou
# step: salt entre elements

llista: list[str] = ["a", "b", "c", "d", "e", "f"]

print(llista[0:3])    # Índexs 0, 1 i 2.
print(llista[2:4])    # Índexs 2 i 3.
print(llista[0:5:2])  # Índexs 0, 2 i 4.
````

#### `enumerate()`

simultàniament:


Això evita haver de gestionar manualment un comptador.

````python
# Exemple amb enumerate()

paraula: list[str] = ["h", "o", "l", "a"]

for index, lletra in enumerate(paraula):
    print(f"índex {index}: lletra {lletra}")

# Exemple amb comptador
print()
for i in range(0, len(paraula)):
    print(f"índex {i}: lletra {paraula[i]}")
````

## ✍️ 1. Exercici: El 8 és primer o és darrer?

Donada una llista de nombres enters, retorna `True` si el nombre `8`
és el primer element o el darrer element de la llista. Retorna `False`
en qualsevol altre cas.

Si la llista és buida, també ha de retornar `False`.

Crida la funció amb quatre llistes diferents:

- dues que retornin `True`;
- dues que retornin `False`.

````python
def primer_darrer_8(llista: list[int]) -> bool:
    """
    Comprova si el nombre 8 és al principi o al final d'una llista.

    Parameters
    ----------
    llista : list[int]
        Llista de nombres enters que es vol comprovar.

    Returns
    -------
    resultat : bool
        True si 8 és el primer o el darrer element de la llista;
        False en qualsevol altre cas.
    """
    buit: bool = len(llista) == 0
    comprova: bool = False

    # Primer la llista no ha de ser buida per poder comprovar el primer i darrer element de la llista
    if not buit:
        principi: int = llista[0]
        final: int = llista[-1]
        comprova: bool = principi == 8 or final == 8
    
    return comprova

llistat1: list[int] = [8,4,2,0]
llistat2: list[int] = [0,2,4,8]
llistat3: list[int] = []
llistat4: list[int] = [1,1,1,1]

print(primer_darrer_8(llistat1))
print(primer_darrer_8(llistat2))
print(primer_darrer_8(llistat3))
print(primer_darrer_8(llistat4))
````

## ✍️ 2. Exercici: A mig camí

Donades dues llistes d'enters, cadascuna de longitud 3, retorna una nova llista que contingui els elements del mig de cada llista.

Per exemple:

- [4, 5, 6], [7, 8, 9] => [5, 8]
- [6, 6, 6], [1, 2, 3] => [6, 2]

````python
def a_mig_cami(
    llista1: list[int],
    llista2: list[int],
) -> list[int]:
    """
    Retorna els elements centrals de dues llistes de longitud tres.

    Parameters
    ----------
    llista1 : list[int]
        Primera llista de tres nombres enters.
    llista2 : list[int]
        Segona llista de tres nombres enters.

    Returns
    -------
    resultat : list[int]
        Llista formada per l'element central de cadascuna de les
        dues llistes.
    """

    # podem definir que l'element central de la llista de 3 valors
    # és la divisió entera de longitud de la llista entre 2 (funciona per les llistes imparelles)
    # ex. 3//2 == 1
    # o bé posar l'índex 1 en aquest cas (perquè només donen llistes de 3 valors)

    mig1 = len(llista1)//2
    mig2 = len(llista2)//2
    
    llistat_mig: list[int] = [llista1[mig1],llista2[mig2]]

    return llistat_mig


print(a_mig_cami([4,5,6],[7,8,9]))
print(a_mig_cami([6,6,6],[1,2,3]))   
````

### Col·lecció `tuple`

Les tuples són seqüències ordenades i **immutables**.

Una vegada creada una tupla, no podem substituir, afegir ni eliminar
directament els seus elements.

La coma és l'element sintàctic que crea la tupla; els parèntesis
s'utilitzen habitualment per fer-ne més clara la representació.

````python
# Creació de tuples

# els ... indiquen que la tupla pot tenir qualsevol nombre d'elements
# en el cas de posar [str, str] estem indicant que només pot tenir 2 elements de tipus str
tupla1: tuple[str, ...] = "a", "b", "c", "d"
tupla2: tuple[str, ...] = ("a", "b", "c", "d")
tupla3: tuple[str, ...] = ("a",)

print(tupla1, tupla2, tupla3)
````

````python
# Consulta de tuples
print(tupla1[0])
print(tupla2[1:3])

# La resta d'operacions de consulta són similars a les de les llistes.
print("* Conversió de llista a tupla")

llista: list[int] = [1, 2, 3]
print(llista)

tupla_convertida: tuple[int, ...] = tuple(llista)
print(tupla_convertida)

print("* Conversió de tupla a llista")

tupla: tuple[int, ...] = (1, 2, 3)
print(tupla)

llista_convertida: list[int] = list(tupla)
print(llista_convertida)
````

````python
# Però atenció, no és possible fer:
tupla5: tuple[int] = (1, 2, 3)
tupla5[1] = 7  #les tuples són immutables
````

## ✍️ 4. Exercici 4 - Nombres entrats per l'usuari

Fes un programa que vagi demanant nombres fins a que l'usuari entri 99 i després els mostri com una tupla.

````python
def tupla_enters_usuari() -> tuple:

    enter: int = 0

    llista_enters: list[int] = []

    while enter != 99:

        enter: int = int(input("Introdueix qualsevol valor (99 per mostrar la tupla): "))

        llista_enters += [enter]

    tupla_enters: tuple[int] = tuple(llista_enters)

    return tupla_enters

tupla_enters_usuari()
        
        
````

### Col·lecció `dict` (diccionari)

Un diccionari associa **claus** amb **valors**.

Les claus han de ser úniques. Els valors, en canvi,
poden ser objectes de qualsevol tipus.

Els diccionaris conserven l'ordre d'inserció dels elements.

````python
# Inicialització de diccionaris

# en aquest cas estem indicant que les claus poden ser str/int
# mentre que els valors només poden ser enters

dicc1: dict[str | int, int] = {}

dicc2: dict[str | int, int] = {
    # la clau "clau1" té associat el valor 1
    # la clau "3" té associat el valor 4
    "clau1": 1,
    3: 4,
}

print(dicc1)
print(dicc2)
````

````python
# Consultar valors d'un diccionari
diccionari: dict[str, int] = {
    "clau1": 1,
    "clau2": 2,
    "clau3": 3,
}

print(diccionari["clau1"])
print(diccionari["clau3"])

print(diccionari.items())   # Parelles clau-valor.
print(diccionari.keys())    # Claus.
print(diccionari.values())  # Valors.

print("clau2" in diccionari)  # Comprova si "clau2" és una clau.
````

````python
# Partint d'aquest diccionari, veurem les operacions bàsiques.
diccionari: dict[str, int] = {
    "clau1": 1,
    "clau2": 2,
    "clau3": 3,
}

print(diccionari)
````

````python
# Afegir
diccionari["clau_nova"] = 9  # afegim una parella clau-valor
print(diccionari)
````

````python
# Modificar
diccionari["clau2"] = 5  # canviem el valor d'una clau existent
print(diccionari)
````

````python
# Esborrar
del diccionari["clau3"]  # esborrem una parella clau-valor existent
print(diccionari)
````

````python
# Esborrar tots els valors
diccionari.clear()  # esborrem tots els valors
print(diccionari)
````

## Complexitat: primera referència

Aquesta secció és només una primera aproximació a la complexitat de les
operacions sobre col·leccions.

Recordem que, per a valors grans de `n`, una operació **O(1)** escala millor
que una **O(log n)**, i aquesta millor que una **O(n)**.

> **Important:** aquestes complexitats descriuen el comportament habitual
> de la implementació CPython. La notació O() descriu com creix el cost
> amb la mida de les dades, no el temps exacte d'execució.

### Llistes (`list`)

Si `n` és la mida de la llista i `k` el nombre d'elements de la part
afectada:

| Operació | Exemple | Complexitat habitual |
|---|---|---:|
| Longitud | `len(llista)` | O(1) |
| Consulta per índex | `llista[i]` | O(1) |
| Modificació per índex | `llista[i] = x` | O(1) |
| Afegir al final | `llista.append(x)` | O(1) amortitzat |
| Eliminar el darrer | `llista.pop()` | O(1) |
| Inserir | `llista.insert(i, x)` | O(n) |
| Eliminar per valor | `llista.remove(x)` | O(n) |
| Cerca | `x in llista` | O(n) |
| Concatenació | `llista1 + llista2` | O(n + m) |
| *Slicing* | `llista[i:j]` | O(k) |
| Còpia | `llista.copy()` | O(n) |
| Comptar | `llista.count(x)` | O(n) |
| Invertir | `llista.reverse()` | O(n) |
| Ordenar | `llista.sort()` | O(n log n) |
| Recorregut | `for x in llista` | O(n) |

`append()` és O(1) **amortitzat**: ocasionalment Python ha de reservar
més memòria, però el cost mitjà d'una seqüència d'operacions continua
sent constant per inserció.

### Tuples (`tuple`)

Les tuples són immutables i, per tant, no disposen d'operacions com
`append()`, `insert()` o `remove()`.

| Operació | Exemple | Complexitat habitual |
|---|---|---:|
| Longitud | `len(tupla)` | O(1) |
| Consulta per índex | `tupla[i]` | O(1) |
| Cerca | `x in tupla` | O(n) |
| Concatenació | `tupla1 + tupla2` | O(n + m) |
| *Slicing* | `tupla[i:j]` | O(k) |
| Comptar | `tupla.count(x)` | O(n) |
| Recorregut | `for x in tupla` | O(n) |

Llistes i tuples tenen sovint la mateixa **complexitat asimptòtica**
per a les operacions de consulta. Que una operació tingui la mateixa
O() no significa que el temps exacte d'execució sigui necessàriament
idèntic.

### Diccionaris (`dict`)

Els diccionaris utilitzen estructures basades en *hashing*. En condicions
habituals, consultar, inserir o eliminar una clau té cost mitjà constant.

| Operació | Exemple | Complexitat mitjana |
|---|---|---:|
| Longitud | `len(dicc)` | O(1) |
| Consulta | `dicc[clau]` | O(1) |
| Pertinença | `clau in dicc` | O(1) |
| Afegir / modificar | `dicc[clau] = valor` | O(1) |
| Eliminar | `del dicc[clau]` | O(1) |
| Crear `keys()` | `dicc.keys()` | O(1) |
| Crear `values()` | `dicc.values()` | O(1) |
| Crear `items()` | `dicc.items()` | O(1) |
| Recórrer claus o valors | `for x in dicc` | O(n) |
| Còpia | `dicc.copy()` | O(n) |
| Buidar | `dicc.clear()` | O(n) |

`keys()`, `values()` i `items()` no creen immediatament una llista amb
tots els elements: retornen **vistes** del diccionari. Crear la vista és
constant; recórrer-ne tots els elements és O(n).

Les operacions de consulta, inserció i eliminació dels diccionaris són
O(1) **de mitjana**. En casos patològics de col·lisions de *hash* poden
tenir un cost pitjor.

### Conjunts (`set`)

Tot i que no els estudiarem detalladament en aquest tema, és útil tenir
una primera referència.

| Operació | Exemple | Complexitat mitjana |
|---|---|---:|
| Longitud | `len(s)` | O(1) |
| Afegir | `s.add(x)` | O(1) |
| Pertinença | `x in s` | O(1) |
| Eliminar | `s.remove(x)` | O(1) |
| Descartar | `s.discard(x)` | O(1) |
| Recorregut | `for x in s` | O(n) |
| Còpia | `s.copy()` | O(n) |
| Unió | `s | t` | O(len(s) + len(t)) |
| Intersecció | `s & t` | O(min(len(s), len(t))) |
| Diferència | `s - t` | O(len(s)) |

Com en els diccionaris, les operacions basades en *hashing* tenen
habitualment complexitat O(1) de mitjana, però això no implica O(1)
en qualsevol situació possible.
