## ✍️ 1. Exercici 1: càlculs pagaments

Ens venen un ordinador per **700 €** i hi hem d'afegir un **21% d'IVA**.

1. Quin serà el preu final?
2. Si l'import final es reparteix entre dos germans, quant haurà de pagar
   cadascun?

Escriu les expressions Python necessàries i mostra els resultats.

````python
def calcul_pagaments(preu: float, percentatge: float) -> None:

    iva: float = percentatge / 100

    preu_final: float = preu*(1 + iva)

    preu_germans: float = preu_final / 2

    print(f"El preu final és {preu_final}")
    print(f"Cada germà haurà de pagar {preu_germans}")

calcul_pagaments (700.00, 21.00)
````

## 🧠 Cadenes de caràcters (`str`)

El text es representa en Python amb el tipus `str`.

Una cadena és una seqüència immutable de caràcters Unicode. Pot contenir,
per exemple, lletres, números, espais, signes de puntuació i caràcters de
molts sistemes d'escriptura.

Podem delimitar una cadena tant amb cometes simples com amb cometes dobles:

```python
"Hola"
'Hola'

````python
# Podem concatenar cadenes amb la suma
'Hola' + "Món" 
````

````python
# Podem repetir cadenes amb el producte
"Qualsevol text " * 10
````

````python
# Podem saber la longitud d'una cadena de caràcters amb la funció len
len("Hola Món")
````

````python
# Podem accedir a un caràcter de la cadena amb la seva posició
# Es comença a comptar per 0
"abcdefghijklmnopqrstuvwxyz"[3]
````

````python
# Podem accedir a una part de la cadena (subcadenes) 
# indicant el caràcter per el que començar i pel que acabar (que no s'inclou). 
# Aquesta transformació es coneix amb el nom "slicing"
"abcdefghijklmnopqrstuvwxyz"[3:5]
````

````python
# Si volem començar des del principi, no cal posar el zero
"abcdefghijklmnopqrstuvwxyz"[:3]
````

````python
# Si volem agafar fins al final, no cal posar el final
"abcdefghijklmnopqrstuvwxyz"[15:]
````

````python
# Podem utilitzar índex negatius per comptar des del final. 
# Per exemple, per agafar el tercer caràcter pel final:
"abcdefghijklmnopqrstuvwxyz"[-3]
````

````python
# O per agafar els tres últims caràcters:
"abcdefghijklmnopqrstuvwxyz"[-3:]
````

````python
# Si no posem ni inici ni fi, ho agafem tot
"abcdefghijklmnopqrstuvwxyz"[:]
````

````python
# Les cadenes tenen alguns mètodes que permeten fer-hi conversions.
# Per exemple, el mètode "upper" transforma la cadena a majúscules
"aBc".upper()
````

````python
# El mètode lower transforma la cadena a minúscules
"ABCd".lower()
````

## ✍️ 5. Exercici 5: Treball amb cadenes i llistes

1. Crea una llista amb els nombres enters de `100` a `110`, ambdós inclosos.
2. A partir de la cadena:

   `123.456.789`

   crea una llista separant-ne els elements pels punts (`.`).

````python
def cadenes_llistes() -> None:

    llista_enters: list[int] = []
    
    for i in range(100, 111):
        
        llista_enters.append(i) # o bé iterar l'enter dins d'una llista llista_enters += [i]

    cadena: str = "123.456.789"

    llista_cadena: list[str] = cadena.split(".")

    print(llista_enters)
    print(llista_cadena)

cadenes_llistes()
        
````
