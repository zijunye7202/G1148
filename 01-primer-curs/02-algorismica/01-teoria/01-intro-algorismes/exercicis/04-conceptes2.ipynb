## 🧠 Assignacions simultànies

### ✍️ Pregunta 1
Executa la següent cella i després canvia els valors. Explica què passa.

```python
x: int = 3
y: int = 4
x = y
y = x

print(x, y) # la variable x assigna el mateix objecte que la variable y
```

### ✍️ Pregunta 2
Executa la següent cella i després canvia els valors. Explica què passa.

```python
x: int = 3
y: int = 4
x, y = y, x

print(x, y) # es tracta d'una assignacio per posicio i desempaquetament
# en aquest cas el valor de de x i y s'intercanvien sense assignar el mateix objecte
# la diferencia esta en que cap dels valors ha sigut modificat previament
# aleshores no tenen el mateix valor a diferencia d'assignar en forma sequencial
```
