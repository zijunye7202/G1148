 ## 🧠 Aquí tens una funció en la que pots veure diversos elements del llenguatge:

+ comentaris,

+ entrada de dades des del teclat,

```python
def comportament_caotic() -> None:
    """Comportament caòtic."""
    print("Aquesta funció implementa un comportament caòtic")
    x: float = float(input("Entra un nombre entre 0 i 1: "))
    for i in range(10):
        x = 3.9 * x * (1 - x)
        print(x)


comportament_caotic()
```

## ✍️ Pregunta 1
Com s'anomena aquesta funció? quines variables es fan servir dins el codi?

Es tracta d'una funcio no retorn, ja que no implementa un return, tal com indica el None.

Dins del codi s'utilitza dos variables:

- x: variable de tipus float o decimal enter, la qual rep un valor a traves de la dada que introdueix l'usuari, tambe s'imprimeix el resultat d'aquesta variable despres d'operar una expressio de forma iterativa

- i: es una variable que rep un valor de forma iterativa, i serveix per recorrer cada element que hi ha en un iterable, en aquest cas un range de 10 nombres.
