## ✍️ 4. Exercici de llistes: Selecció per posicions

Escriu una funció que, donada una llista, retorni una tupla amb **tres
llistes**:

1. una amb els elements situats a les posicions `0, 3, 6, ...`;
2. una altra amb els elements situats a les posicions `1, 4, 7, ...`;
3. una tercera amb els elements situats a les posicions `2, 5, 8, ...`.

No pots fer cap iteració.

Exemple:

```python
llista = [
    "a", 1, True,
    "b", 2, False,
    "c", 3, None,
]

resultat = separar_posicions(llista)
print(resultat)
```

Resultat:

```text
(
    ["a", "b", "c"],
    [1, 2, 3],
    [True, False, None],
)
```

*Atenció*: la llista pot contenir qualssevol valors.

Font: Adaptat de [Python praktikum / Alessandro Mammana](https://www.molgen.mpg.de/3372238/python_praesentation.pdf)

````python
# Solució exercici 3


def separar_posicions(
    llista: list[object],
) -> tuple[list[object], list[object], list[object]]:
    """
    Separa els elements d'una llista en tres grups segons la seva posició.

    Parameters
    ----------
    llista : list[object]
        Llista que es vol separar.

    Returns
    -------
    resultat : tuple[list[object], list[object], list[object]]
        Tres llistes amb les posicions 0, 3, 6, ...;
        1, 4, 7, ...; i 2, 5, 8, ...
    """
    return llista[::3], llista[1::3], llista[2::3]


resultat: tuple[
    list[object],
    list[object],
    list[object],
] = separar_posicions(
    [
        "a", 1, True,
        "b", 2, False,
        "c", 3, None,
    ]
)

print(resultat)
````

## ✍️ 5. Exercici de llistes, matrius

Escriu una funció que, donats:

- un nombre `n_files`;
- un nombre `n_columnes`;
- un valor `valor`;

retorni una matriu amb `n_files` files i `n_columnes` columnes en què
totes les posicions continguin `valor`.

Representa la matriu com una llista de llistes.

Cada fila ha de ser una llista independent de les altres.

Exemple:

```python
matriu = generar_matriu(2, 4, 7)
print(matriu)
```

Resultat:

```text
[
    [7, 7, 7, 7],
    [7, 7, 7, 7],
]
```

Font: Adaptat de [Python praktikum / Alessandro Mammana](https://www.molgen.mpg.de/3372238/python_praesentation.pdf)

````python
# Solució exercici 4


def generar_matriu(
    n_files: int,
    n_columnes: int,
    valor: object,
) -> list[list[object]]:
    """
    Genera una matriu inicialitzada amb un valor determinat.

    Parameters
    ----------
    n_files : int
        Nombre de files.
    n_columnes : int
        Nombre de columnes.
    valor : object
        Valor amb què s'inicialitzen les posicions de la matriu.

    Returns
    -------
    matriu : list[list[object]]
        Matriu amb les dimensions i el valor indicats.
    """
    matriu: list[list[object]] = []

    for _ in range(n_files):
        fila: list[object] = []

        for _ in range(n_columnes):
            fila.append(valor)

        matriu.append(fila)

    return matriu


matriu: list[list[object]] = generar_matriu(2, 4, 7)
print(matriu)
````

## ✍️ 56. Exercici de diccionaris: Invertir un diccionari

Escriu una funció que, donat un diccionari, en retorni un altre amb les
claus i els valors intercanviats.

Exemple:

```python
invertir_dic(
    {
        "name": "ale",
        "age": "impolite to ask",
        "nfingers": "20",
    }
)
```

ha de retornar:

```python
{
    "ale": "name",
    "impolite to ask": "age",
    "20": "nfingers",
}
```

Per a aquest exercici pots assumir que:

- tots els valors poden utilitzar-se com a claus d'un diccionari;
- no hi ha valors repetits.

Pensa què passaria si dos elements del diccionari original tinguessin
el mateix valor.

Font: Adaptat de [Python praktikum / Alessandro Mammana](https://www.molgen.mpg.de/3372238/python_praesentation.pdf)

````python
# Solució exercici 5

from typing import TypeVar

K = TypeVar("K")
V = TypeVar("V")


def invertir_dic(dic: dict[K, V]) -> dict[V, K]:
    """
    Inverteix les claus i els valors d'un diccionari.

    Parameters
    ----------
    dic : dict[K, V]
        Diccionari que es vol invertir.

    Returns
    -------
    resultat : dict[V, K]
        Diccionari amb les claus i els valors intercanviats.
    """
    nou_diccionari: dict[V, K] = {}

    for clau, valor in dic.items():
        nou_diccionari[valor] = clau

    return nou_diccionari


def invertir_dic2(dic: dict[K, V]) -> dict[V, K]:
    """
    Inverteix un diccionari utilitzant keys() i accés per clau.

    Parameters
    ----------
    dic : dict[K, V]
        Diccionari que es vol invertir.

    Returns
    -------
    resultat : dict[V, K]
        Diccionari amb les claus i els valors intercanviats.
    """
    nou_diccionari: dict[V, K] = {}

    for clau in dic.keys():
        valor: V = dic[clau]
        nou_diccionari[valor] = clau

    return nou_diccionari
````

````python
diccionari: dict[str, str] = {
    "name": "ale",
    "age": "impolite to ask",
    "nfingers": "20",
}

invertir_dic(diccionari)
````

## ✍️ 7. Exercici de diccionaris: Aminoàcids

Els aminoàcids es codifiquen mitjançant seqüències de codons, formats per
tres bases. Cada codó codifica un únic aminoàcid, però un mateix aminoàcid
pot correspondre a més d'un codó.

Escriu una funció `codificar()` que rebi:

1. una llista de codons sense duplicats;
2. una llista amb els aminoàcids corresponents, en el mateix ordre;
3. una seqüència de codons que es vol traduir.

La funció ha de retornar la seqüència d'aminoàcids corresponent.

Exemple:

```python
codificar(
    ["TTT", "TTC", "ATT", "GCG"],
    ["Phe", "Phe", "Ile", "Ala"],
    ["ATT", "GCG", "ATT", "GCG", "TTT", "ATT", "GCG"],
)
```

ha de retornar:

```python
["Ile", "Ala", "Ile", "Ala", "Phe", "Ile", "Ala"]
```

Pots provar també les crides següents:

```python
codificar(
    ["AAA", "CCC", "GGG"],
    ["Lys", "Pro", "Gly"],
    ["GGG", "AAA", "CCC", "CCC", "GGG"],
)

codificar(
    ["UCU", "UCC", "UCA", "UCG"],
    ["Ser", "Ser", "Ser", "Ser"],
    ["UCG", "UCU", "UCA", "UCC", "UCU"],
)

codificar(
    ["AUG", "UUU", "GAA", "CCU"],
    ["Met", "Phe", "Glu", "Pro"],
    ["AUG", "UUU", "GAA", "AUG", "AUG", "CCU", "GAA"],
)

codificar(
    ["ACG"],
    ["Thr"],
    ["ACG", "ACG", "ACG"],
)
```

Font: Adaptat de [Python praktikum / Alessandro Mammana](https://www.molgen.mpg.de/3372238/python_praesentation.pdf)

````python
# Solució exercici 6


def codificar(
    codons: list[str],
    aminoacids: list[str],
    sequencia: list[str],
) -> list[str]:
    """
    Tradueix una seqüència de codons als aminoàcids corresponents.

    Parameters
    ----------
    codons : list[str]
        Llista de codons únics.
    aminoacids : list[str]
        Aminoàcids corresponents als codons.
    sequencia : list[str]
        Seqüència de codons que es vol traduir.

    Returns
    -------
    resultat : list[str]
        Seqüència d'aminoàcids corresponent.
    """
    correspondencies: dict[str, str] = dict(
        zip(codons, aminoacids, strict=True)
    )

    return [
        correspondencies[codon]
        for codon in sequencia
    ]
````

````python
codificar(
    ["TTT", "TTC", "ATT", "GCG"],
    ["Phe", "Phe", "Ile", "Ala"],
    ["ATT", "GCG", "ATT", "GCG", "TTT", "ATT", "GCG"],
)
````

## ✍️ 8. Exercici de diccionaris: Freqüències

Escriu una funció que llegeixi un fitxer de text i calculi la freqüència
de les lletres.

La funció ha de:

1. convertir el text a minúscules;
2. considerar només les lletres `a-z`;
3. ignorar espais, dígits, puntuació i qualsevol altre caràcter;
4. retornar les lletres i les seves freqüències en ordre descendent de
   freqüència.

Prova la funció amb textos en diferents idiomes i observa si les
freqüències varien.

Pots comparar els resultats amb les taules de l'article de la
[Viquipèdia sobre freqüència de les lletres](https://en.wikipedia.org/wiki/Letter_frequency).

Font: Adaptat de [Python for Everybody / Charles Severance](https://books.trinket.io/pfe/index.html)

````python
# Solució exercici 7

from collections import Counter


def ordre_descendent(
    ruta_fitxer: str = "lletra.txt",
) -> list[tuple[int, str]]:
    """
    Calcula les freqüències de les lletres d'un fitxer.

    Parameters
    ----------
    ruta_fitxer : str
        Ruta del fitxer de text que es vol analitzar.

    Returns
    -------
    resultat : list[tuple[int, str]]
        Llista de tuples (freqüència, lletra), ordenada de més a menys
        freqüència.
    """
    lletres: str = "abcdefghijklmnopqrstuvwxyz"

    # Llegir el contingut del fitxer.
    with open(
        ruta_fitxer,
        "r",
        encoding="utf-8",
    ) as fitxer:
        text: str = fitxer.read().lower()

    # Comptar només les lletres.
    comptador: Counter[str] = Counter(
        caracter
        for caracter in text
        if caracter in lletres
    )

    # Ordenar per freqüència (descendent).
    tuples_ordenades: list[tuple[int, str]] = sorted(
        (
            (frequencia, lletra)
            for lletra, frequencia in comptador.items()
        ),
        reverse=True,
    )

    return tuples_ordenades
````

````python
# Prova de l'exercici 7

from pathlib import Path

fitxer_prova: Path = Path("lletra_prova.txt")

fitxer_prova.write_text(
    "AAA bb c! 123 à",
    encoding="utf-8",
)

resultat: list[tuple[int, str]] = ordre_descendent(
    "lletra_prova.txt"
)

assert resultat == [(3, "a"), (2, "b"), (1, "c")]
````
