## ✍️ Problema 1: Num. enregistraments

Calcula quants enregistraments s'han fet durant les  dades indicades? (1 a 3 de maig, 1 enregistrament per minut, un únic sensor).

```python
enregistraments: int = 3 * 24 * 60

print(f"{enregistraments} enregistraments")
```

Fes una funció que calculi el nombre d'enregistraments a partir del nombre de sensors indicats i dels dies enregistrats.

```python
def num_enregistraments() -> None:

    sensors: int = int(input("Introdueix el nombre de sensors: "))
    dies: int = int(input("Introdueix el nombre de dies: "))
    
    enregistrament_diari: int = 1 * 24 * 60

    enregistraments_totals: int = enregistrament_diari * sensors * dies

    print(f"Per a {sensors} sensors en {dies} dies hi ha {enregistraments_totals} enregistraments.")


num_enregistraments()
```

```python
# separa les dades de la primera linia, mostrant text del tipus de dada

def llegeix_primera_fila() -> None:

    with open(
        '9808Dies1a3reduit.csv',
        'r',
        encoding = 'utf-8',
    ) as fitxer:

        linia1: str = (fitxer.readlines())[0]

        linia1_separada = linia1.split(';')

        print(f" El dia es: {linia1_separada[0]}")
        print(f" El temps es: {linia1_separada[1]}")
        print(f" El nivell de soroll es: {float(linia1_separada[2])/10}")

llegeix_primera_fila()
```

```python
# fes el mateix per a cada linia
# despres mostra nomes el nivell de soroll
# finalment mostra el valor maxim del nivell de soroll

def llegir_dades_fitxer() -> None:

    with open(
        '9808Dies1a3reduit.csv',
        'r',
        encoding = 'utf-8',
    ) as fitxer:

        max: int = 0

        for linia in fitxer:

            llista_dades: list[str] = linia.split(';')

            dia: int = int(llista_dades[0])
            temps: str = llista_dades[1]
            nivell_soroll: int = int(llista_dades[2])

            print(f" El dia es: {dia}")
            print(f" El temps es: {temps}")
            print(f" El nivell de soroll es: {nivell_soroll}")
            print()

            # ara farem un bucle for per trobar el nivell de soroll maxim
            
            if nivell_soroll > max:

                max = nivell_soroll

        print(f"El nivell de soroll maxim es: {max}")

llegir_dades_fitxer()
```

## ✍️ Problema 3: Màxima contaminació

Pas 1. Començant amb  el fitxer reduit, fes una funció que et retorni la dada de màxim volum de soroll.

Pas 2. Ara fes que la funció també et retorni l'hora en què s'ha produït el màxim volum.

Pas 3. Aplica la funció al fitxer més gran.

```python
def maxima_contaminacio() -> float | str:

    with open(
        '9808Dies1a3.csv',
        'r',
        encoding = 'utf-8',
    ) as fitxer:

        # definim un maxim volum per tal de comparar de forma iterativa
        # aixi podem comparar amb cada nivell de soroll del fitxer

        max: float = 0
        temps_max: str = ""
        
        for linia in fitxer:

            linia_separada: list[str] = linia.split(';')

            #dividim entre 10 per passar a decibels
            temps: str = linia_separada[1]
            nivell_soroll: float = float(linia_separada[2])/10
            
            if nivell_soroll > max:

                max = nivell_soroll
                temps_max = temps
                
        print(f"El nivell de soroll maxim va ser de {max} decibels a les {temps_max}.")

        return max, temps_max

maxima_contaminacio()
```

## ✍️ Problema 4: Infraccions dels nivells tolerats

Tot i que aquest sensor es troba en una zona no sensible (C), els nivells de soroll admesos sovint s'excedeixen. Podem suposar que els veïns tenen una tolerància de +15 els valors indicats.

Revisa els continguts del fitxer gran i mostra per pantalla totes les hores en que s'ha sobrepassat el nivell de soroll acceptat. 

```python
# nivell de soroll acceptat + tolerancia en la zona C
""" 
Dia: de 7h a 22:59h
Nit: de 23:00h a 6:59h

Zona Dia Nit
C	 85	 75
"""

# cal classificar cada registre segons dia i nit
# com que esta en format HH:MM:SS llavors els hem de separar-les per classificar

# ficare en dues llistes dia i nit aquests aquells enregistraments

def temps_contaminacio() -> None:

    with open(
        '9808Dies1a3.csv',
        'r',
        encoding = 'utf-8',
    ) as fitxer:

        soroll_dia: float = 85.00
        soroll_nit: float = 75.00

        for linia in fitxer:

            dades: list[str] = linia.split(';')

            temps: str = dades[1]
            nivell_soroll:float = float(dades[2])/10

            format_temps: list[str] = temps.split(':')
            
            hores: int = int(format_temps[0])
                
            if hores >= 7 and hores <= 22:
                
                if nivell_soroll > soroll_dia:
                
                    print(f"Al mati, sobre les {temps} es va superar els {soroll_dia} decibels")

            else: 
                
                if nivell_soroll > soroll_nit:

                    print(f"Al la nit, sobre les {temps} es va superar els {soroll_nit} decibels")

temps_contaminacio()
```
