<p class="exercici">Exercici 1: LLista de zeros</p>

Escriu una funció que donat un nombre $n$ crei una llista amb n zeros. No pots fer cap iteració.

```python
def zeros(n: int) -> list[int]:
    """
    Crea una llista formada per n zeros sense utilitzar iteracions.

    Parameters
    ----------
    n : int
        Nombre de zeros que ha de contenir la llista.

    Returns
    -------
    resultat : list[int]
        Llista formada per n zeros.
    """
    return [0] * n


zeros(4)
```

<p class="exercici">Exercici 2: LLista selectiva</p>

Escriu una funció que donat un nombre $n$ i una llista, retorni una segona llista amb els elements de la llista inicials que són estrictament menors al nombre $n$ donat. 

```python
def llista_selectiva(llista: list[int], n: int) -> list[int]:
    """
    Retorna els elements d'una llista que són estrictament menors que n.

    Parameters
    ----------
    llista : list[int]
        Llista de nombres enters que es vol filtrar.
    n : int
        Valor límit per seleccionar els elements.

    Returns
    -------
    resultat : list[int]
        Elements de la llista original que són menors que n.
    """
    nova_llista: list[int] = []

    for valor in llista:
        if valor < n:
            nova_llista.append(valor)

    return nova_llista


llista_selectiva([1, 15, 5, 3, 2, 12, 99, 8, 6], 10)
```

<p class="exercici">Exercici 3: LLista selectiva ordenada</p>

Aquest exercici requereix coneixements d'*ordenació i cerca*.

Escriu una funció que donat un nombre $n$ i una llista ordenada, retorni una segona llista amb els elements de la llista inicials que són estrictament menors al nombre $n$ donat. 

```python
def cerca_binaria_posicio(
    n: int,
    llista: list[int],
    inici: int,
    final: int,
) -> int:
    """
    Troba la primera posició d'una llista ordenada amb un valor >= n.

    Utilitza una cerca binària recursiva sobre l'interval
    semiobert [inici, final).

    Parameters
    ----------
    n : int
        Valor que es vol localitzar.
    llista : list[int]
        Llista ordenada de nombres enters.
    inici : int
        Índex inicial de l'interval de cerca.
    final : int
        Índex posterior al final de l'interval de cerca.

    Returns
    -------
    posicio : int
        Primera posició que conté un valor major o igual que n.
    """
    if inici >= final:
        return inici

    mig: int = (inici + final) // 2

    if llista[mig] < n:
        return cerca_binaria_posicio(n, llista, mig + 1, final)

    return cerca_binaria_posicio(n, llista, inici, mig)


def llista_selectiva_ordenada(
    llista: list[int],
    n: int,
) -> list[int]:
    """
    Retorna els elements d'una llista ordenada estrictament menors que n.

    Parameters
    ----------
    llista : list[int]
        Llista ordenada de nombres enters.
    n : int
        Valor límit per seleccionar els elements.

    Returns
    -------
    resultat : list[int]
        Elements de la llista que són estrictament menors que n.
    """
    posicio: int = cerca_binaria_posicio(
        n,
        llista,
        0,
        len(llista),
    )

    return llista[:posicio]


llista_selectiva_ordenada([10, 11, 12], 10)
```

## ✍️ 1. Exercici 4: Separació d'elements alterns

Escriu una funció que donada una llista retorni una tupla amb dues llistes: una amb els elements en posicions senars de la llista i l'altra amb els elements en posicions parelles de la llista. No pots fer cap iteració.

```python
def separar_alterns(
    llista: list[object],
) -> tuple[list[object], list[object]]:
    """
    Separa una llista en dues llistes amb els elements de posicions alternes.

    Parameters
    ----------
    llista : list[object]
        Llista que es vol separar.

    Returns
    -------
    resultat : tuple[list[object], list[object]]
        Tupla formada pels elements de les posicions 0, 2, 4, ...
        i pels elements de les posicions 1, 3, 5, ...
    """
    return llista[::2], llista[1::2]


separar_alterns(
    ["un", 1, "dos", 2, "tres", 3, "quatre", 4]
)
```

<p class="exercici">Exercici 5: Matriu</p>

Escriu una funció que donats dos nombres $nfiles$ i $ncolumnes$ retorni una matriu de zeros amb nfiles files i ncolumnes columnes. Cal crear la matriu com una llista de llistes.

```python
def generar_matriu(
    n_files: int,
    n_columnes: int,
) -> list[list[int]]:
    """
    Genera una matriu de zeros mitjançant dos bucles.

    Parameters
    ----------
    n_files : int
        Nombre de files de la matriu.
    n_columnes : int
        Nombre de columnes de la matriu.

    Returns
    -------
    matriu : list[list[int]]
        Matriu de zeros amb les dimensions indicades.
    """
    matriu: list[list[int]] = []

    for _ in range(n_files):
        fila: list[int] = []

        for _ in range(n_columnes):
            fila.append(0)

        matriu.append(fila)

    return matriu


def generar_matriu_simplificada(
    n_files: int,
    n_columnes: int,
) -> list[list[int]]:
    """
    Genera una matriu de zeros creant cada fila per repetició.

    Parameters
    ----------
    n_files : int
        Nombre de files de la matriu.
    n_columnes : int
        Nombre de columnes de la matriu.

    Returns
    -------
    matriu : list[list[int]]
        Matriu de zeros amb les dimensions indicades.
    """
    matriu: list[list[int]] = []

    for _ in range(n_files):
        matriu.append([0] * n_columnes)

    return matriu
```

```python
matriu1: list[list[int]] = generar_matriu(2, 4)
print(matriu1)
```

```python
matriu2: list[list[int]] = generar_matriu_simplificada(2, 4)
print(matriu2)
```

<p class="exercici">Exercici 6: Entrada en seqüència i càlculs sobre llistes</p>

Escriu un programa que demani un número a l'usuari repetidament fins que aquest introdueixi -99. Quan hagi acabat d'introduir els números mostra per pantalla la suma dels nombres introduits, quants nombres s'han introduit i la mitjana aritmètica.

```python
from statistics import mean


def calcular_estadistiques() -> None:
    """
    Llegeix nombres enters i en calcula la suma, quantitat i mitjana.

    La introducció de -99 indica que s'ha acabat la seqüència.

    Parameters
    ----------
    No té paràmetres.

    Returns
    -------
    None
        La funció mostra els resultats per pantalla.
    """
    nombres: list[int] = []

    numero: int = int(
        input("Entra un número. -99 per acabar: ")
    )

    while numero != -99:
        nombres.append(numero)
        numero = int(
            input("Entra un número. -99 per acabar: ")
        )

    if not nombres:
        print("No s'ha introduït cap nombre.")
        return

    suma: int = sum(nombres)
    quantitat: int = len(nombres)
    mitjana: int | float = mean(nombres)

    print(
        f"S'han introduït {quantitat} nombres, "
        f"amb una suma total de {suma} "
        f"i una mitjana aritmètica de {mitjana}."
    )


calcular_estadistiques()
```

<p class="exercici">Exercici 7: Entrada en seqüència i càlculs sobre llistes. Variant</p>

Escriu un programa que demani un número a l'usuari repetidament fins que aquest introdueixi -99. Ves recopilant els nombres en una llista. Quan hagi acabat d'introduir els números mostra per pantalla el mínim i el màxim de la llista.

```python
def calcular_extrems() -> None:
    """
    Llegeix nombres enters i mostra el mínim i el màxim introduïts.

    La introducció de -99 indica que s'ha acabat la seqüència.

    Parameters
    ----------
    No té paràmetres.

    Returns
    -------
    None
        La funció mostra el mínim i el màxim per pantalla.
    """
    nombres: list[int] = []

    numero: int = int(
        input("Entra un número. -99 per acabar: ")
    )

    while numero != -99:
        nombres.append(numero)
        numero = int(
            input("Entra un número. -99 per acabar: ")
        )

    if not nombres:
        print("No s'ha introduït cap nombre.")
        return

    maxim: int = max(nombres)
    minim: int = min(nombres)

    print(
        f"El número màxim introduït és {maxim} "
        f"i el mínim és {minim}."
    )


calcular_extrems()
```

## ✍️ 2. Exercici 9: Ordre invers de les paraules

Escriu una funció que donada una frase la retorni a l'usuari amb les paraules en ordre invers.

```python
def ordre_invers(frase: str) -> str:
    """
    Retorna les paraules d'una frase en ordre invers.

    Parameters
    ----------
    frase : str
        Frase que es vol invertir.

    Returns
    -------
    resultat : str
        Frase amb les paraules en ordre invers.
    """
    paraules: list[str] = frase.split()
    return " ".join(paraules[::-1])


ordre_invers("Em dic Andreu")
```
