## ✍️ 1. Exercici 1: Tallar cadena

Escriu una funció que donats una cadena i un nombre n<<len(cadena), elimini els primers n caràcters i els darrers n caràcters de la cadena.

```python
def tallar_cadena(cadena: str, n: int) -> str:
    """
    Elimina els primers i els darrers n caràcters d'una cadena.

    Parameters
    ----------
    cadena : str
        Cadena que es vol retallar.
    n : int
        Nombre de caràcters que s'han d'eliminar de cada extrem.

    Returns
    -------
    resultat : str
        Cadena resultant després d'eliminar els caràcters dels extrems.
    """
    return cadena[n : len(cadena) - n]
```

```python
tallar_cadena("algorísmica", 2)
```

## ✍️ 3. Exercici 3: Generador de passwords

Escriu una funció que generi passwords. La generació ha de ser aleatòria i ha de generar un password nou a cada execució. L'usuari ha de determinar la llargada del password.

```python
import secrets


def generar_password(longitud: int) -> str:
    """
    Genera una contrasenya aleatòria amb la longitud indicada.

    Parameters
    ----------
    longitud : int
        Nombre de caràcters que ha de tenir la contrasenya.

    Returns
    -------
    password : str
        Contrasenya generada aleatòriament.
    """
    caracters: str = (
        "abcdefghijklmnopqrstuvwxyz"
        "01234567890"
        "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
        "!@#$%^&*()?"
    )

    password: str = "".join(
        secrets.choice(caracters) for _ in range(longitud)
    )

    return password
```

```python
generar_password(8)
```
