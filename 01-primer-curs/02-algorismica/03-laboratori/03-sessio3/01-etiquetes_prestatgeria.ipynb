## ✍️ Etiquetes alfabètiques de prestatgeria

Un magatzem etiqueta les prestatgeries amb la seqüència `A, B, ..., Z, AA, AB, ...`.Quina etiqueta tindrà la prestatgeria número 12?

### ⚠️ Conversió entre números i càrcters
Abans de començar recordem que A = chr(65) i Z = chr(90), i ord("A") = 65, i ord("Z") = 90.

```python
print(chr(65))
print(chr(90))
print(ord("A"))
print(ord("Z"))
```

### 🧠 Tasca


```python
def etiqueta_prestatgeria(numero: int) -> str:
    """
    Aquesta funció diu quina etiqueta li correspon
    a una prestatgeria donada.

    Parameters
    ----------
    num: int

    Returns
    prestatgeria: str
    """

    etiqueta: str = ""

    # la prestatgeria es una col·lecció A-Z de base 26
    # cada lletra és una posició obtinguda del numero % 26 ex. A = 26 % 26 = 0
    # amb aixo podem transformar cada lletra A = 65 + residu (65+1 = B, 65+2 = C, etc...)
    # per a valors més grans llavors hem de dividir de forma iterativa entre 26 per tornar a afegir les lletres restants
    # com que començem per 1 aleshores descomptem 1 per el valor A agafat

    while numero > 0:

        numero -= 1

        resta: int = numero % 26

        lletra: str = chr(ord('A') + resta)

        etiqueta += lletra

        numero //= 26

    return etiqueta

etiqueta_prestatgeria(703)        
```

### ✅ Asserts públics


```python
assert etiqueta_prestatgeria(1) == 'A'
assert etiqueta_prestatgeria(26) == 'Z'
assert etiqueta_prestatgeria(27) == 'AA'
assert etiqueta_prestatgeria(54) == 'BB'
assert etiqueta_prestatgeria(703) == 'AAA'
```
