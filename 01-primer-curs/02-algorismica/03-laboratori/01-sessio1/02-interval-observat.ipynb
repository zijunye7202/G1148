## ✍️ Interval observat 

Una sèrie de mesures enteres s'ha registrat en una llista. Retorna una tupla amb el menor i el major valor observats, no pots usar cap funció de Python que resolgui el quid de la qüestió. La solució ha de ser lineal.


### 🧠 Tasca


```python
def interval_observat(mesures: list[int]) -> tuple[int, int]:
    """Retorna el mínim i el màxim de la llista."""

    minim: int = mesures[0]
    maxim: int = mesures[0]

    for i in mesures: 
    
        if i > maxim:
    
            maxim = i
    
        elif i < minim:
    
            minim = i

    tupla: tuple = (minim, maxim)

    return tupla

interval_observat([12, -3, 8, 21, 4])
```

### ✅ Asserts públics


```python
assert interval_observat([12, -3, 8, 21, 4]) == (-3, 21)
assert interval_observat([7]) == (7, 7)
assert interval_observat([5, 5, 5, 5]) == (5, 5)
```
