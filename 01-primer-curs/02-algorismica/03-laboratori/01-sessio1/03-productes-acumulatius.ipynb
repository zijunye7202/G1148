## ✍️ 2.8B Productes acumulats consecutius


Considera la successió que comença per `1` i en què cada terme nou s'obté multiplicant l'anterior pel següent enter positiu: `1, 2, 6, 24, 120, ...`.

Donat un límit `L`, retorna tots els termes estrictament menors que `L`. No utilitzis cap funció que resolgui el quid de la qüestió.


### 🧠 Tasca


```python
def productes_acumulats(lim: int) -> list[int]:
    """Retorna els termes de la successió estrictament menors que lim."""

    # n = 1, a = 1
    # a = 1 x 1 = 1, n = 2
    # a = 1 x 2 = 2, n = 3
    # a = 2 x 3 = 6, n = 4
    # a = 6 x 4 = 24, n = 5
    # a = 24 x 5 = 120, n = 6
    
    # sempre segueix el mateix patro: a = n * a on n += 1 a cada iteracio
    # llavors donada la condicio de que n * a sigui menor que L podem fer un bucle while
    # actualitzar el valor d'a i fer una suma acumulativa d'n

    L: int = lim
    n: int = 1
    a: int = 1
    
    llista_prod: list[int] = []
    
    while (n * a) < L:
        
        llista_prod.append(n*a)
        a: int = n*a
        n += 1

    return llista_prod

productes_acumulats(25)
```

### ✅ Asserts públics


```python
assert productes_acumulats(25) == [1, 2, 6, 24]
assert productes_acumulats(7) == [1, 2, 6]
assert productes_acumulats(2) == [1]
```
