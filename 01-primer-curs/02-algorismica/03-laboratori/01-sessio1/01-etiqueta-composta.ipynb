## ✍️ Clau d’una etiqueta composta


Una etiqueta està formada per paraules separades per espais i no conté signes de puntuació ni números. La seva **clau curta** es forma concatenant, en majúscules, la primera lletra de cada paraula.

Per exemple, `centre de dades obertes` produeix `CDDO`.


```python
def clau_etiqueta(text: str) -> str:

    clau_curta: str = ""
    
    clau_separada: list[str] = text.split()

    for terme in clau_separada:

        clau_curta += terme[0].upper()

    return clau_curta
    
clau_etiqueta("centre de dades obertes")
```

### 🧠 Tasca


```python
def clau_etiqueta(text: str) -> str:
    """Construeix la clau curta d'una etiqueta composta."""

    llista = text.split() # Creo una llista que conté termes, dels quals els espais blancs son el separador

    acronim = "" # Afegeixo totes les inicials de cada paraula dins d'aquesta cadena buida, sent aquesta el valor str que vull retornar

    for index in range (0, len(llista)): # Vull récorrer cada terme dins la llista, passant per tots els índexs existents
        paraula = llista[index] # Definim la paraula com l'element que es troba a cada un dels índexs que hi ha dins la llista
        inicial = paraula[0].upper() # Agafem la inicial en majúscula de la paraula corresponent, que es troba en el primer índex
        acronim = acronim + inicial # Cada cop que agafem una inicial majúscula la sumem dins de la cadena creada

    return acronim # Retornem la cadena que conté totes les inicials majúscules com a valor de retorn

clau1 = clau_etiqueta("centre de dades obertes")
clau2 = clau_etiqueta("laboratori sistemes interactius")
clau3 = clau_etiqueta("Python")

print(clau1)
print(clau2)
print(clau3)
```

### ✅ Asserts públics


```python
assert clau_etiqueta('centre de dades obertes') == 'CDDO'
assert clau_etiqueta('laboratori sistemes interactius') == 'LSI'
assert clau_etiqueta('Python') == 'P'
```
