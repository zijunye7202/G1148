## ✍️ Codis de lectura simètrica

Un codi és **simètric** si es llegeix igual d'esquerra a dreta que de dreta a esquerra. Implementa aquesta comprovació sense dependre d'una funció específica de Python per invertir la cadena.

Després, determina si algun desplaçament circular del codi és simètric. Un desplaçament circular mou el primer caràcter al final; repetint l'operació s'obtenen totes les rotacions possibles.


### 🧠 Tasca


```python
def codi_simmetric(codi: str) -> bool:
    """Indica si el codi és simètric."""

    # comparem la primera meitat amb la segona recorrent la cadena inversament 
    return (codi[:len(codi)//2] == codi[:len(codi)//2-1:-1])
    
def alguna_rotacio_simetrica(codi: str) -> bool:
    """Indica si alguna rotació circular del codi és simètrica."""

    # de forma iterativa canviem de posicio de la primera lletra
    # i afegim a una llista de booleans si es simetric el codi
    # la rotacio sera simetrica si hi ha algun True dins la llista
    
    rotacio: list[bool] = []
    
    for i in range(len(codi)):

        codi = codi[1:]+codi[0] # AABB: ABBA, BBAA, BAAB, AABB
        rotacio.append(codi_simmetric(codi))

    # print(rotacio)

    return (True in rotacio)
# print(codi_simmetric('ABCCBA'))
# print(codi_simmetric('ABCA'))
# print(codi_simmetric('X'))
# print(alguna_rotacio_simetrica('AABB'))
# print(alguna_rotacio_simetrica('ABCD'))
```

### ✅ Asserts públics


```python
assert codi_simmetric('ABCCBA') is True
assert codi_simmetric('ABCA') is False
assert codi_simmetric('X') is True
assert alguna_rotacio_simetrica('AABB') is True
assert alguna_rotacio_simetrica('ABCD') is False
```

### 📊 Complexitat

Indica i justifica la complexitat temporal i espacial de la solució.


La solucio te una complexitat d'ordre O(n) i O(n^2):

- De cara a la primera funcio de codi simmetric, com que estem fent slicings per tal de retornar un boolea que compara cadenes aleshores l'ordre es n, ja que a mesura que augmentem n elements la complexitat es proporcional.

- Respecte a la complexitat de la segona funcio, aqui interve creacio de llista, len(codi), i el metode append() que son d'ordre O(1), o be quan fem iteracions (ordre n). De cara al slicing i concatenacio de cadenes, o quan cridem la funcio de codi simmetric, i quan fem retorn del boolea (el qual itera la llista de booleans), tenim funcions d'ordre O(n) tambe, ja que fem recorreguts on la complexitat es proporcional als elements n d'entrada. Aleshores tenint en compte que aquestes instruccions es recorren de forma iterativa llavors tenim un programa formada per un proces iteratiu n que multiplica per les intruccions interiors d'ordre n, per tant, es d'ordre n**2.

```python
# Complexitat temporal i mitjana de les mesures temporals

import time
import statistics

inici: float = time.perf_counter()

def codi_simmetric(codi: str) -> bool:
    """Indica si el codi és simètric."""

    # comparem la primera meitat amb la segona recorrent la cadena inversament 
    return (codi[:len(codi)//2] == codi[:len(codi)//2-1:-1])
    
def alguna_rotacio_simetrica(codi: str) -> bool:
    """Indica si alguna rotació circular del codi és simètrica."""

    # de forma iterativa canviem de posicio de la primera lletra
    # i afegim a una llista de booleans si es simetric el codi
    # la rotacio sera simetrica si hi ha algun True dins la llista
    
    rotacio: list[bool] = []
    
    for i in range(len(codi)):

        codi = codi[1:]+codi[0] # AABB: ABBA, BBAA, BAAB, AABB
        rotacio.append(codi_simmetric(codi))

    # print(rotacio)

    return (True in rotacio)

llista_t: list[int] = []
for i in range(1000):
    temps: float = time.perf_counter() - inici
    llista_t += [temps] 

print(statistics.mean(llista_t))
```

## ⭐ Repte

Retorna la primera rotació simètrica trobada. Si no n’hi ha cap, retorna una cadena buida.


```python
def codi_simetric(codi:str) -> bool:

    meitat: int = len(codi)//2
    return (codi[:meitat] == codi[:meitat-1:-1])

def primera_rotacio_simetrica(codi: str) -> str:
    """Retorna la primera rotació simètrica o ''."""
    
    for i in range(0, len(codi)):

        codi = codi[1:]+codi[0]

        if codi_simetric(codi):

            return codi

        else:

            return ''
    
print(primera_rotacio_simetrica('AABB') == 'ABBA')
print(primera_rotacio_simetrica('ABCD') == '')
```

### ✅ Asserts públics del repte


```python
assert primera_rotacio_simetrica('AABB') == 'ABBA'
assert primera_rotacio_simetrica('ABCD') == ''
```
