## ✍️ Valors amb restriccions simultànies

Resol dues cerques acotades.

**A.** Troba el valor `n` més gran que compleixi simultàniament:

- `n < 50000`;
- `n` és un quadrat perfecte;
- existeix un enter `k` tal que `n = 4 * k! + 4`.

**B.** Un bloc de procés dura 720 segons. Troba tots els nombres de blocs `b` per als quals `b * 720 = k!` per algun `k < 11`.


### 🧠 Tasca


```python
import math

# n < 50000
# (isqrt(n))**2 == n
# n = (4 * k! + 4) < 50000

def maxim_restringit() -> int:
    
    """Troba el valor màxim que compleix les tres condicions."""

    maxim: int = 0

    for n in range (50000):

        if (math.isqrt(n))**2 == n:

            k: int = 0

            while (4 * math.factorial(k) + 4) < n:

                k += 1

            if (4 * math.factorial(k) + 4) == n and n > maxim:

                maxim = n
    
    # print(maxim)
    return maxim
    
print(maxim_restringit() == 20164)

def blocs_factorials() -> list[int]:
    """Retorna els nombres de blocs que corresponen exactament a un factorial."""

    # Un bloc de procés dura 720 segons. Troba tots els nombres de blocs b per als quals b * 720 = k! per algun k < 11.

    # k < 11
    # b * 720 = k!
    # b = k!/720

    # llista_b: list[int] = []
    
    # for k in range(11):

    #     b: int = (math.factorial(k))//720

    #     if b * 720 == math.factorial(k):
            
    #         llista_b.append(b)

    # return llista_b

    # donat que a = b * q + r
    # a = math.factorial(k) = k!
    # q = a // b = bloc b
    # b = 720
    # r = math.factorial(k) % b
    # si r = 0 llavors retornem el bloc b (q)

    return [
        (math.factorial(k)//720)
        for k in range(11)
        if (math.factorial(k) % 720) == 0
    ]
                
blocs_factorials()
    
```

### ✅ Asserts públics


```python
assert maxim_restringit() == 20164
assert blocs_factorials() == [1, 7, 56, 504, 5040]
```
