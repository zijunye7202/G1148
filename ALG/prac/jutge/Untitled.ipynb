```python
a: int = int(input())

print("YES") if ((a % 400 == 0) or (a % 4 == 0 and a % 100 != 0)) else print("NO")
```

```python
a1, a2, b1, b2 = map(int, input().split())

if (a1 <= b2 and a2 >= b1):

    print(f"[{max(a1, b1)},{min(a2, b2)}]")

else:
    print("[]")
```

```python
a1, a2, b1, b2 = map(int, input().split())

if (a1 > b1 and a2 < b2):
    print("1")

elif (b1 > a1 and b2 < a2):
    print("2")

elif (a1 == b1 and a2 == b2):
    print("=")

else: 
    print("?") 
```

```python
a1, a2, b1, b2 = map(int, input().split())

if (a1 == b1 and a2 == b2):
    a = "="

elif (a1 >= b1 and a2 <= b2):
    a = "1"

elif (a1 <= b1 and a2 >= b2):
    a = "2"

else:
    a = "?"
    
if (a1 <= b2 and a2 >= b1):

    print(f"{a},[{max(a1, b1)}, {min(a2, b2)}]")

else:

    print(f"{a},[]")
```
