```python
# funció que llegeix un nombre en una base donada i retorna en decimal

def base_decimal (nombre: int, base: int) -> int:

    # iterem el nombre transformat en string, aixi trobem el nombre de digits
    # la base decimal ve donada com la suma dels productes de base b per cada digít
    # la base b depèn de la seva posició (de més a menys)
    # tal que la suma és de d*b^(x) on x = digits - 1 i es va reduint

    nombres: str = str(nombre) 
    digits: int = len(nombres)
    x: int = digits - 1
    b: int = base
    decimal: int = 0

    for i in nombres:

        d: int = int(i)

        decimal += (d * (b ** x))

        x -= 1

    return decimal

print(base_decimal(10110, 2))
```

```python
# teorema de Fermat

# explica que donat un nombre enter p primer, per a qualsevol enter a entre 1 i p - 1
# es compleix que a^(p-1) % p = 1

import random

def fermat(num: int, test_count: int) -> bool: #volem saber si és primer

    if num == 1: # p no pot menor o igual a 1

        return False

    for x in range(test_count): 

        val: int = random.randint(1, num - 1) # 1 <= a < p - 1
        
        if pow(val, num-1) % num != 1:

            return False

    return True

print(fermat(7, 1))    
```

```python
# divisió aritmètica

# donat divisió x/y
# sabem que x = y * q + r

import math

def divisio(x: int, y: int) -> list[int]:

    # volem retornar q, r = [7, 1] de forma recursiva 
    # llavors començarem amb q, r = [0,0]

    if x <= 0:

        return 0, 0

    if y == 1: 

        return x, 0

    q, r = divisio(x//2, y) # de forma recursiva dividim x fins obtenir 0, y

    # ara desfem la recursivitat

    q *= 2 
    r *= 2

    if x % 2 != 0:

        r += 1

    if r >= y:

        r -= y
        q += 1

    return [q] + [r]

print(divisio(22, 3))
```

```python
# funció calculadora, donat 2 enters i operació, retorna el resultat 
# aplicant algorisme rus i divisió

def calculadora() -> int | list[int]: 

    x: int = int(input("Introdueix un valor enter x: "))
    y: int = int(input("Introdueix un valor enter y: "))

    op: str = input("Introdueix una operació (multiplicació o divisió): ")

    if op.lower() in ("divisió", "divisio"):

        if x < 0 or y <= 0:

                print("Cal que x >= 0 i que y > 0")
                return -1

        def divisio(x: int, y: int) -> list[int]:

            if x == 0:

                return [0,0]

            if y == 1:

                return [x, 0]

            q, r = divisio(x//2, y)

            q *= 2;
            r *= 2;

            if x % 2 != 0:

                r += 1

            if r >= y:

                r -= y  
                q += 1

            return [q] + [r]

        return divisio(x, y)
    
    elif op.lower() in ("multiplicació", "multiplicacio"):

        resultat: int = 0

        # operem en valor absolut per després retornar amb el signe indicat
        signe: int = -1 if x < 0 else 1
        x = abs(x)

        while x > 0:

            if x % 2 != 0:

                resultat += y

            x //= 2
            y *= 2

        return [signe * resultat]

    else: 

        print("Operació no vàlida!")
        return -1

print(calculadora())
```
