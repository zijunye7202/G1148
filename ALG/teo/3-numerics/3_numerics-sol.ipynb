## Exercici 1

Copia el codi de les solucions del tema 1 demanar_números i factorial, i calcula’n els passos computacionals i la complexitat O gran.


```python
def demanar_numeros() -> None:
    """
    Demana números a l'usuari fins que entri el 100.
    Mostra un missatge segons el rang del número:
      - Negatiu (<0)
      - Molt petit (>=0 i <50)
      - Molt gran (>=50)
    Quan s'entra 100, mostra totes les desenes de 0 a 100 i acaba.
    """
    num: int = 0-------------------------------------------------1 pas  

    while num != 100:--------------------------------------------El de dintre x tants passos com entrades de l'usuari
        num = int(input("Entra un número (100 per acabar): "))---1 pas conversió, 1 pas assignació

        if num < 0:----------------------------------------------1 pas
            print("Has entrat un número negatiu.")---------------1 pas
        elif 0 <= num < 50:--------------------------------------2 passos
            print("Has entrat un número molt petit.")------------1 pas
        else:
            print("Has entrat un número molt gran.")-------------1 pas

    # Quan num és 100, mostrar les desenes
    for i in range(0, 101, 10):---------------------------------El de dintre x 10
        print(i)------------------------------------------------1 pass

El cost de l'algorisme és O(n) on la n és el nombre de números que entra l'usuari.
```

```python
def factorial(num: int) -> int:
    """
    Calcula el factorial d'un número enter no negatiu.

    :param num: Número enter >= 0.
    :return: Factorial de num.
    """
    resultat: int = 1-----------------------1 pas
    for i in range(2, num + 1):-------------El de dintre x n-1 pas, on n és el valor de num
        resultat *= i-----------------------n2 (és una multiplicació i pot ser de nombres grans, la n podria no correspondre a num)
    return resultat-------------------------1 pass

La complexitat és O(n3), tot i que el resultat pot creixer molt ràpid i ser més costós.
```

## Exercici 2

Escriu una funció per llegeixi un nombre en una base donada. L’entrada serà el nombre i la base, i la sortida el seu valor en decimal.

```python
def llegir_en_decimal(nombre:int, base:int) -> int:
    """
    Calcula el valor decimal de nombre.

    :param 
       nombre: Número en base base
       base: La base actual del nombre
    :return: 
       El valor de nombre en decimal
    """
    nombrestr:str = str(nombre)
    potencia:int = len(nombrestr)-1
    valor:int = 0
    for i in nombrestr:
        digit = int(i)
        valor += digit * base**potencia
        potencia = potencia - 1
    return valor
```

```python
assert llegir_en_decimal(11,2) == 3
assert llegir_en_decimal(11,16) == 17
```

## Exercici 4

Fes una implementació pròpia de la funció pow de Python.


```python
def pow_custom(base:int, exponent:int, modulus:int) -> int:
    """
    Implementació pròpia de pow(base, exponent, modulus) amb exponenciació per quadrats.

    :param base: nombre base
    :param exponent: exponent (≥ 0)
    :param modulus: mòdul opcional
    :return: base^exponent (si modulus no es dóna), o (base^exponent) % modulus
    """
    result:int = 1
    if modulus > 1:
        base = base % modulus
    

    e = exponent
    while e > 0:
        if e % 2 == 1:  # si l'exponent és senar
            result = result * base
            if modulus > 1:
                result %= modulus
        base = base * base
        if modulus > 1:
            base %= modulus
        e //= 2

    return result
```

```python
print(pow_custom(2, 10, 1))      # 1024
```

```python
print(pow_custom(2, 10, 1000))   # 24
```
