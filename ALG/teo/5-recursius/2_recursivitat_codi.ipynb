## 🧠 Càlcul del factorial d'un nombre donat

$$
n! =
\begin{cases}
1 & \text{si } n = 0, \\
1 & \text{si } n = 1, \\
n \cdot (n-1)! & \text{si } n > 1
\end{cases}
$$

```python
def factorial(n: int) -> int:
    if n == 0:
        return 1
    return n * factorial(n - 1)
```

$$
F(n) =
\begin{cases}
0 & \text{si } n = 0, \\
1 & \text{si } n = 1, \\
F(n-1) + F(n-2) & \text{si } n > 1
\end{cases}
$$

```python
def fibonacci(n: int) -> int:
    """
    Calcula el n-èssim nombre de Fibonacci de forma recursiva.

    :param n: Índex (n ≥ 0) del nombre de Fibonacci a calcular.
    :return: El nombre de Fibonacci a la posició n.
    """
    if n==0:
        return n
    if n==1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

```python
from typing import List

def fibonacci_iteratiu(n: int) -> int:
    """
    Calcula el n-èssim nombre de Fibonacci de forma iterativa.

    :param n: Índex (n ≥ 0) del nombre de Fibonacci a calcular.
    :return: El nombre de Fibonacci a la posició n.
    """
    if n == 0:
        return 0

    seq: List[int] = [0, 1]
    for i in range(2, n + 1):
        seq.append(seq[i - 1] + seq[i - 2])

    return seq[n]
```

```python
def fibonacci_iteratiu_optimitzat(n: int) -> int:
    """
    Calcula el n-èssim nombre de Fibonacci de forma iterativa
    utilitzant només dues variables (optimitzat en memòria).

    :param n: Índex (n ≥ 0) del nombre de Fibonacci a calcular.
    :return: El nombre de Fibonacci a la posició n.
    """
    a: int = 0
    b: int = 1

    for i in range(1, n + 1):
        a, b = b, a + b

    return a
```

```python
def factorial_iteratiu(n: int) -> int:
    resultat = 1
    for i in range(1, n + 1):
        resultat *= i
    return resultat
```
