# <span class="tema">Recursivitat</span><br/> 📘 Comparació de les tres versions de Fibonacci

En aquest notebook es reproduiran les tres versions vistes a teoria de Fibonacci i per cadascuna es mostraran les crides que es fan, i es calcularà el cost i l'ordre de complexitat O gran.

## 🧠 Versió recursiva

````python
def fibonacci(n: int) -> int:
    """
    Calcula el n-èssim nombre de Fibonacci de forma recursiva.

    :param n: Índex (n ≥ 0) del nombre de Fibonacci a calcular.
    :return: El nombre de Fibonacci a la posició n.
    """
    print("Estem cridant la funció amb el valor ",n)
    if n==0:
        return n
    if n==1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)
````

````python
fibonacci(5)
````

### ⏱️ Càlcul del cost

Si n <= 1 fem una crida, el cost és constant O(1)

si n>1 la funció fa dues crides més un cost constant (comparacions i suma)
$T(n)=T(n−1)+T(n−2)+C$ i això és repetirà fins arribar al cas base. Això ho podem representar amb un arbre **binari**, en el que cada fulla de l’arbre de crides és un **cas base** (`n = 0` o `n = 1`).

I quantes fulles hi ha? el nombre de fulles és aproximadament **F(n)** (nombre de Fibonacci), que creix com:
  $$
  F(n) \approx \frac{\varphi^n}{\sqrt{5}}
  $$
  on $\varphi \approx 1.618$.

El cost de l'algorisme és proporcional al nombre total de nodes de l’arbre.
- En un arbre binari ple:
  $$
  \text{Nodes totals} \approx 2 \times \text{Fulles}
  $$
- Si treiem les constants multiplicatives, en notació Big-O, ens queda un cost de:
  $$
  T(n) = O(\varphi^n) \quad \text{aproximadament } O(2^n)
  $$

I a banda hem gastat molta memòria perquè la pila d'execució ha tingut també totes aquestes crides.


## 🧠 Versió iterativa

````python
from typing import List

def fibonacci_iteratiu(n: int) -> int:
    """
    Calcula el n-èssim nombre de Fibonacci de forma iterativa.

    :param n: Índex (n ≥ 0) del nombre de Fibonacci a calcular.
    :return: El nombre de Fibonacci a la posició n.
    """
    if n == 0:
        return 0

    seq: List[int] = [0, 1]
    print(seq)
    for i in range(2, n + 1):
        seq.append(seq[i - 1] + seq[i - 2])
        print(seq)

    return seq[n]
````

````python
fibonacci_iteratiu(5)
````

### ⏱️ Càlcul del cost

```python
def fibonacci_iteratiu(n: int) -> int:
    if n == 0:-----------------------------------O(1)
        return 0---------------------------------O(1)

    seq: List[int] = [0, 1]----------------------2 passos, O(1)
    for i in range(2, n + 1):--------------------el de dintre per n - 1
        seq.append(seq[i - 1] + seq[i - 2])------2  passos, O(1)
        --------el cost de la iteració és (n - 1) x 2 => O(2n)
    return seq[n]--------------------------------O(1) no fa una còpia, retorna per referència
```

    En total el cost és O(1 + 2n + 1) => O(2n) => O(n)

    Aquests càlculs són vàlids mentre els nombres són menors a 18446744073709551616 (2^64). Però cal tenir en compte que F(94) ja és superior a aquest valor.

```python
def fibonacci_iteratiu_optimitzat(n: int) -> int:
    a: int = 0-----------------------------------O(1)
    b: int = 1-----------------------------------O(1)

    for i in range(1, n + 1):--------------------el de dintre per n
        a, b = b, a + b--------------------------2 passos, O(1)
    --------el cost de la iteració és (n) x 2 => O(2n)
    return a-------------------------------------O(1)
```

    En total el cost és O(1 + 2n + 1) => O(2n) => O(n)
