# <span class="tema">Recursivitat</span><br/> 📘Exemple de funció que es pot expressar amb recursivitat

En aquest notebook es reproduiran les tres versions vistes a teoria de Fibonacci i per cadascuna es mostraran les crides que es fan, i es calcularà el cost i l'ordre de complexitat O gran.

## 🧠 Versió recursiva

````python
def suma_llista_recursiva(valors: list[int]) -> int:
    """
    Calcula la suma dels elements d'una llista utilitzant recursivitat.

    :param valors: Llista d'enters a sumar.
    :return: La suma dels elements de la llista.
    """
    if len(valors) == 0:
        return 0
    return valors[0] + suma_llista_recursiva(valors[1:])
````

````python
suma_llista_recursiva([4,8,1,4,-3,9])
````

````python
def suma_llista_iterativa(valors: list[int]) -> int:
    """
    Calcula la suma dels elements d'una llista de manera iterativa.

    :param valors: Llista d'enters a sumar.
    :return: La suma dels elements de la llista.
    """
    suma: int = 0
    for element in valors:
        suma += element
    return suma
````

````python
suma_llista_iterativa([4,8,1,4,-3,9])
````

### ⏱️ Càlcul del cost

En els dos casos el cost és O(n) però en la versió recursiva tenim una pila d'execució molt més gran.

```python
def suma_llista(valors: List[int]) -> int:
    if len(valors) == 0:------------------------------O(1)
        return 0--------------------------------------O(1)
    return valors[0] + suma_llista(valors[1:])--------1 pas accés + 1 pas suma + crida
```
Quantes crides es fan? n-1. Cada crida O(1). En total el cost és O(n)

A memòria hi ha tota la pila d'execució, una entrada per cada crida (n entrades)


```python
def suma_llista_iterativa(valors: List[int]) -> int:
    suma: int = 0 ------------------------------O(1)
    for element in valors:----------------------el de dintre per n, on n és el nombre d'elements de la llista
        suma += element-------------------------O(1)
    return suma---------------------------------O(1)
```

En total el cost és O(1 + n + 1) => O(n)

A memòria només hi ha la pila d'una única funció.
