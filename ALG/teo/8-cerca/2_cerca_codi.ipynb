## 🔍➡️ Cerca líneal

```python
def cerca_lineal(A:list[any],K:any):
  i:int=0
  while i < len(A) and A[i] != K:
    i += 1
  if i < len(A):
     return i
  else:
    return -1
```

```python
cerca_lineal([23,27,5,90,12,44,38,84,77],38)
```

```python
def cerca_lineal(A:list[any],K:any):
  A.append(K)
  i:int=0
  while A[i] != K:
    i += 1
  if i < len(A)-1:
     return i
  else:
    return -1
```

## 🔎➗ Cerca binària

```python
def cerca_binaria_rec(A:list[any],K:any):
     mig = len(A) // 2
     if A[mig] > K:
          return cerca_binaria_rec(A[0:mig], K)
     elif A[mig] < K: 
          return cerca_binaria_rec(A[mig+1:len(A)], K)+mig+1 # sumo el desplaçament
     else: 
          return mig                           
     return -1
```

```python
A = [3,14,27,31,39,42,55,70,74,81,85,93,98]
cerca_binaria_rec(A,81)
```

```python
def cerca_binaria(A:list[any],K:any):
    inici = 0
    final = len(A)-1
    while inici <= final:
       mig = (inici + final) // 2
       if A[mig] > K:
            final = mig - 1
       elif A[mig] < K: 
            inici = mig + 1
       else: 
           return mig                           
    return -1
```

```python
A = [3,14,27,31,39,42,55,70,74,81,85,93,98]
cerca_binaria(A,70)
```

## 🤯⚠️ Cerca exhaustiva, exemple

```python
import math

def func1d(x:int): # funció multimodal de la que buscarem max
    y: float = x * math.sin(10*math.pi*(x))+1.0
    return y

def frange(start: float, stop: float, step: float): 
# definim un range sobre floats per conveniència 
    current: float = start 
    while current < stop:
        yield current   # yield proveeix un resultat i segueix
        current += step

def cerca_exhaustiva_func1d():   # la funció de cerca 
    x:int = -1  #valor inicial d'x
    maxim:float = func1d(x)  # valor inicial d'y 
    for i in frange(-1.0, 2.0, 0.01):
        if func1d(i)>maxim: 
            maxim = func1d(i)
            x = i
    print(x, maxim)
```

## 🤯⚠️ Cerca aleatòria, exemple

```python
import random

def cerca_aleatoria_func1d():
    x = -1  #valor inicial d'x
    maxim=func1d(x)  # valor inicial d'y 
    for i in range(1000): #provem 1000 valors aleatòriament
      xtemp = (random.random()*3.0)-1.0   # valors entre -1 i 2
      if func1d(xtemp)>maxim: 
          maxim=func1d(xtemp)
          x=xtemp
    return x, maxim
```
