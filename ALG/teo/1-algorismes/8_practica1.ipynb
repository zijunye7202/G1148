## ✍️ 1. Completar funcions

````python
## Completa les funcions següents segons les indicacions donades.
# Després crida-les i comprova que funcionen.


def mostrar_nom(nom: str) -> None:  # falta el nom de la funció
    """
    Mostra un nom per pantalla.

    Parameters
    ----------
    nom : str
        Nom que es vol mostrar.

    Returns
    -------
    None
        La funció mostra el nom per pantalla.
    """
    print(nom)

def calcul(x: int, y: int) -> None:
    """
    Fa una operació amb dos nombres enters i en mostra el resultat.

    Parameters
    ----------
    x : int
        Primer nombre.
    y : int
        Segon nombre.

    Returns
    -------
    None
        La funció mostra el resultat per pantalla.
    """
    # Tria l'operació que s'ha de fer amb x i y.

    print(x + y)

def calcul_per_teclat() -> None:
    """
    Demana dos nombres enters i en mostra el resultat d'un càlcul.

    Returns
    -------
    None
        La funció mostra el resultat per pantalla.
    """
    x: int = int(input("Digues un nombre: "))
    y: int = int(input("Digues un altre nombre: ")) # completa-ho
    # Demana un segon nombre i guarda'l a y.
    # Després mostra el producte de x i y.
    print (f"El producte de {x} i {y} és {x * y}")
    
def elevat_a_3(n: int) -> None:
    """
    Mostra un nombre elevat al cub.

    Parameters
    ----------
    n : int
        Nombre que es vol elevar al cub.

    Returns
    -------
    None
        La funció mostra el resultat per pantalla.
    """
    print (n**3) # Com es feia la potència?
````

````python
# aquesta estrategia no es la mes optima d'emplear, es mes recomanable fer assignacio per posicio i desempaquetament
# ja que fer-ho amb bucles implica que el valor de la variable anterior sempre sigui modificat per l'anterior d'aquesta

def intercanvi_valors(x: int, y: int, w: int, z: int) -> None:

    # vull mostrar el contingut de forma iterativa aleshores faig l'assignacio amb llistes
    llista_valors: list[int] = [x, y, w, z]
    
    llista_valors = [z, x, y, w]

    variables: list[str] = ['x','y','w','z']
    
    for j in range(0, len(llista_valors)):
        print(f" {variables[j]} = {llista_valors[j]}")

intercanvi_valors(4,5,6,7)


        
````
