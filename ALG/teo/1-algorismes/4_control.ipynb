## Introducció

Les estructures de control ens permeten alterar la seqüència d'execució de les instruccions d'un programa. Sense elles els programes s'executarien sempre línia a línia.

## Condicionals: `if`, `elif`, `else`

Les estructures condicionals permeten decidir quin codi s'ha d'executar
en funció d'una o més condicions.

Python avalua primer la condició de l'`if`:

- si és `True`, executa el bloc corresponent;
- si és `False`, passa al primer `elif`, si n'hi ha;
- pot haver-hi zero, un o diversos blocs `elif`;
- si cap condició és certa i existeix un `else`, s'executa aquest darrer bloc.

Els blocs de codi es defineixen mitjançant la **indentació**.

Exemple:

```python
if condicio1:
    instruccions
elif condicio2:
    instruccions
else:
    instruccions

````python
# Exemple
a: int = 5
if a > 4:
    print("a és major que 4")
elif a > 2:
    print("a és major que 2 i igual o menor que 4")
else:
    print("a és igual o menor que 2")

# Prova el mateix exemple canviant el valor d'a per a = 3 i per a = 1
````

## ✍️ 1. Exercici 1: comparació de dues sumes

Escriu una acció que, donats quatre nombres, compari la suma dels dos
primers amb la suma dels dos darrers.

Ha de mostrar un missatge diferent en cadascun dels casos següents:

1. la suma dels dos primers és més gran;
2. la suma dels dos darrers és més gran;
3. les dues sumes són iguals.

Prova l'acció amb tres casos que permetin verificar les tres situacions.

````python
def comparacio_sumes(a: int, b: int, c: int, d: int) -> None:

    suma_primers: int = a + b
    suma_darrers: int = c + d

    if (suma_primers > suma_darrers):
        print("la suma dels dos primers és més gran")
    elif (suma_primers < suma_darrers):
        print("la suma dels dos darrers és més gran")
    elif (suma_primers == suma_darrers):
        print("les dues sumes són iguals")

comparacio_sumes(1,2,3,4)
comparacio_sumes(4,3,2,1)
comparacio_sumes(1,1,1,1)
````

## Iteracions: `for` i `while`

Les iteracions permeten repetir un conjunt d'instruccions.

Python disposa principalment de dues estructures d'iteració:

- **`while`**: repeteix un bloc mentre una condició sigui certa;
- **`for`**: recorre successivament els elements d'un objecte iterable.

Utilitzarem habitualment `while` quan la finalització depengui d'una
condició que pot variar durant l'execució.

Utilitzarem habitualment `for` quan vulguem recórrer una seqüència,
una col·lecció o un rang de valors.

````python
# Exemple
a: int = 1
while a < 5:
    print(a)
    a += 1  # En Python l'operació d'autoincrement s'escriu així

# Prova el mateix exemple canviant el valor de la condició per 3 i del valor inicial d'a per 4
````

````python
# Exemple2
a: int = 3
while a < 13:
    print(a)
    a += 3  # En Python l'operació d'autoincrement s'escriu així
````

````python
# Exemple3: atenció abans d'executar-lo pensa què farà!
a: int = 3
while a < 13:
    print(a) # mostrarà a de manera indefinida ja que sempre es compleix la condició
````

### Iteració `while`

Una iteració `while` repeteix un bloc d'instruccions **mentre la seva
condició sigui `True`**.

Per tant, és important assegurar-nos que durant l'execució hi hagi alguna
possibilitat que la condició arribi a ser `False`. Això pot passar, per
exemple:

- perquè modifiquem una variable dins del bucle;
- perquè rebem una nova entrada de l'usuari;
- perquè trobem el valor que estàvem buscant.

En cas contrari podem crear un **bucle infinit**.

El `while` és especialment útil quan no sabem a priori quantes iteracions
seran necessàries.

````python
nombre: int = int(input("Introdueix un nombre menor que 5: "))

while nombre >= 5:
    print("El nombre introduït no és menor que 5.")
    nombre = int(input("Introdueix un nombre menor que 5: "))

print(f"Molt bé, {nombre} és menor que 5.")
# prova diverses execucions del codi
````

### Iteració `for`

La instrucció `for` permet recórrer els elements d'un **iterable**.

Un iterable pot ser, per exemple:

- un `range`;
- una llista;
- una tupla;
- una cadena;
- un diccionari;
- altres objectes que es poden recórrer.

Per exemple:

```python
for animal in ["gat", "gos", "canari"]:
    print(animal)

Quan volem generar una seqüència de nombres enters utilitzem habitualment
range():

```Python
range(start, stop, step)
```

on:

+ *start* és el primer valor;
+ *stop* és el límit final, que no s'inclou;
+ *step* és l'increment entre dos valors consecutius.

Si no indiquem start, el valor per defecte és 0.

Si no indiquem step, el valor per defecte és 1.

````python
# Exemple 1
for i in range(0, 10):  # start: 0, stop: 10, step: 1 per defecte
    print(i)
````
