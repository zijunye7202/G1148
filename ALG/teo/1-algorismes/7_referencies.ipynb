# 📘 Python - Referències

A Python, les variables no contenen directament els objectes: són **noms
associats a objectes**.

La funció incorporada `id()` retorna un enter que identifica de manera
única un objecte durant la seva vida.

Dos noms associats al mateix objecte tindran el mateix `id()`.

> No hem d'interpretar `id()` com una adreça de memòria. La documentació
> de Python només garanteix que aquest identificador sigui únic i constant
> durant la vida de l'objecte.

```python
a: str = "banana"
b: str = "banana"

c: list[int] = [1, 2, 3]
d: list[int] = [1, 2, 3]

e: list[int] = c  # e i c fan referència al mateix objecte
```

```python
id(a)
```

```python
id(b)
```

```python
id(c)
```

```python
id(d)
```

```python
id(e)
```

```python
c is e
```

En el cas de `c` i `d`, hem creat dues llistes diferents amb el mateix
contingut. Per tant, són objectes diferents i tenen identitats diferents.

En canvi:

`e = c`

no crea una nova llista. Associa el nom `e` al mateix objecte al qual ja
està associat `c`. Per això `c` i `e` tenen el mateix `id()`.

Amb alguns objectes immutables, com determinades cadenes o nombres, Python
pot reutilitzar internament un mateix objecte com a optimització. Per
aquest motiu no hem de deduir les regles d'identitat comparant literals.

Per comprovar si dos noms fan referència al mateix objecte podem utilitzar
l'operador `is`.

Per comprovar si dos objectes tenen el mateix valor utilitzem `==`.

### Assignació i còpia de col·leccions

Una assignació com:

`b = a`

**no copia la llista**. Simplement associa un segon nom al mateix objecte.
Diem que `a` i `b` són dos **àlies** del mateix objecte.

Per tant, si modifiquem la llista mitjançant un dels dos noms, el canvi
també és visible des de l'altre.

Si volem crear una nova llista amb els mateixos elements podem fer una
**còpia superficial** (*shallow copy*), per exemple:

`c = a.copy()`

o:

`c = a[:]`

La nova llista és un objecte diferent, però els seus elements continuen
sent referències als mateixos objectes que contenia la llista original.

```python
# Efectes de tenir dos noms associats a la mateixa llista
a: list[int] = [1, 2, 3]
print("a original:", a)

b: list[int] = a  # b és un àlies d'a

a.append(4)

print("a modificada:", a)
print("b:", b)

# Amb una còpia superficial obtenim una nova llista
a = [1, 2, 3]
print("a original:", a)

c: list[int] = a.copy()

a.append(4)

print("a modificada:", a)
print("c:", c)
```

La còpia que hem fet amb `copy()` és una **còpia superficial** (*shallow
copy*).

Això significa que es crea una nova llista, però no es copien recursivament
els objectes continguts dins seu.

Aquest detall és especialment important quan una llista conté altres
objectes mutables, com ara altres llistes.

```python
def afegeix_element(llista: list[str]) -> None:
    """
    Afegeix un element a una llista.

    Parameters
    ----------
    llista : list[str]
        Llista que es vol modificar.

    Returns
    -------
    None
        La funció modifica la llista rebuda.
    """
    llista.append("nou")


llista: list[str] = ["un", "un altre"]

afegeix_element(llista)

print(llista)
```

```python
# Amb una còpia evitem alterar la llista original.


def afegeix_element_a_copia(llista: list[str]) -> list[str]:
    """
    Afegeix un element a una còpia de la llista.

    Parameters
    ----------
    llista : list[str]
        Llista original, que no es vol modificar.

    Returns
    -------
    copia : list[str]
        Nova llista amb l'element afegit.
    """
    copia: list[str] = llista.copy()
    copia.append("nou")

    return copia


llista: list[str] = ["un", "un altre"]

nova_llista: list[str] = afegeix_element_a_copia(llista)

print("Nova llista:", nova_llista)
print("Llista original:", llista)
```
