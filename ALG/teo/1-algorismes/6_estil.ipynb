## Docstrings i comentaris

### Docstrings

Tota funció ha de començar amb una **docstring** que expliqui:

- què fa la funció;
- els seus paràmetres i tipus;
- què representa cada paràmetre;
- el valor retornat i el seu tipus.

En aquesta assignatura utilitzarem un format inspirat en l'estil
**NumPy**.

```python
def comptar_sumar(n: int) -> int:
    """
    Calcula la suma dels nombres enters entre 1 i n - 1.

    A més, mostra per pantalla els nombres que es van sumant.

    Parameters
    ----------
    n : int
        Límit superior de la seqüència. El valor n no s'inclou.

    Returns
    -------
    suma : int
        Suma dels nombres enters entre 1 i n - 1.
    """
    suma: int = 0

    for i in range(1, n):
        suma += i
        print(i)

    return suma
```

Una docstring no és el mateix que un comentari. Forma part de la
documentació de la funció i es pot consultar, per exemple, amb `help()`.

### Comentaris

Els comentaris breus s'escriuen amb `#`:

```python
temperatura: float = 18.5

# Convertim la temperatura de Celsius a Fahrenheit.
fahrenheit: float = temperatura * 9 / 5 + 32
```

Els comentaris han d'explicar aspectes que **no siguin evidents a partir
del mateix codi**. No cal comentar cada instrucció.

## Noms de funcions i variables

Els noms de funcions i variables s'escriuen en minúscules.

Quan estan formats per diverses paraules, les separem amb guions baixos
(`snake_case`):

```python
edat: int = 25
meu_cotxe: str = "Toyota"
valor_aproximat: float = 3.567


def calcular_suma(a: int, b: int) -> int:
    return a + b
```

En aquesta assignatura utilitzarem, com a convenció, identificadors sense
accents ni altres signes diacrítics:

```python
numero_maxim
calcul_mitjana
```

i evitarem:

```text
número_màxim
càlcul_mitjana
```

Aquesta és una **convenció de l'assignatura** per facilitar la portabilitat
i la consistència del codi; Python permet identificadors Unicode.

## Constants

Les constants s'escriuen habitualment en majúscules, separant les paraules
amb guions baixos:

```python
MAXIM: int = 9_999_999
TIPUS_CANVI: float = 0.86
```

## Anotacions de tipus

En aquesta assignatura indicarem explícitament els tipus dels paràmetres,
els retorns i les variables quan sigui útil.

Utilitzarem la sintaxi moderna de Python:

```python
nom: str = "Anna"
edat: int = 25
notes: list[float] = [7.5, 8.0, 9.5]
coordenades: tuple[float, float] = (2.0, 3.5)
telefon: str | None = None
```

Les funcions indicaran també els tipus dels paràmetres i del retorn:

```python
def multiplicar(a: int, b: int) -> int:
    return a * b
```

Si una funció no retorna cap valor utilitzarem:

```python
def mostrar_missatge(missatge: str) -> None:
    print(missatge)
```

## Indentació

Utilitzarem **4 espais** per cada nivell d'indentació.

```python
def mostrar_nombres() -> None:
    for i in range(2):
        print("Nou bloc")
        print(i)
```

No barrejarem tabuladors i espais.

## Longitud de les línies

Segons PEP 8, el codi hauria de tenir com a màxim **79 caràcters per
línia**. Els comentaris i docstrings es recomana limitar-los a
**72 caràcters per línia**.

Per dividir una expressió llarga, és preferible aprofitar la continuació
implícita dins de parèntesis, claudàtors o claus.

Per exemple:

```python
print(
    "Aquest text és massa llarg per escriure'l",
    "còmodament en una única línia.",
)
```

També podem dividir una expressió:

```python
resultat: float = (
    valor_inicial
    + increment
    - descompte
)
```

En general, és preferible aquesta forma abans que utilitzar `\` per
continuar explícitament una línia.

## Línies en blanc

Les funcions i classes definides al nivell principal del programa se
separen amb **dues línies en blanc**.

Dins d'una funció podem utilitzar línies en blanc, amb moderació, per
separar blocs lògics.

```python
def funcio1() -> None:
    pass


def funcio2() -> None:
    pass
```

## Espais en blanc

Deixarem un espai a cada costat dels operadors:

```python
suma: int = a + b
resultat: float = valor * 2.5
```

Després d'una coma deixarem un espai:

```python
llista: list[str] = ["a", "b", "c"]
```

En les anotacions de tipus:

```python
edat: int = 25
nom: str = "Anna"
```

No deixarem espais al final de les línies.

## Imports

Els `import` se situen habitualment al principi del programa.

Cada mòdul s'importa en una línia diferent:

```python
import math
import string
```

Els imports s'agrupen habitualment en aquest ordre:

1. biblioteca estàndard de Python;
2. biblioteques de tercers;
3. mòduls propis del projecte.

Es deixa una línia en blanc entre els diferents grups.

Per exemple:

```python
import math
from pathlib import Path

import pandas as pd

from projecte import utilitats
```

## Principi general

L'objectiu de les normes d'estil no és només que el programa
«compleixi unes regles», sinó que el codi sigui:

- fàcil de llegir;
- fàcil d'entendre;
- consistent;
- fàcil de modificar i mantenir.
