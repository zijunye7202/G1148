# 📘 Python: Entrada (Input) i Sortida (Output)

## Entrada de dades: `input()`

La funció `input()` permet demanar dades a l'usuari.

La sintaxi bàsica és:

```python
variable = input("Text que es mostra a l'usuari: ")
```

`input()` mostra el text indicat, espera que l'usuari introdueixi una dada
i retorna el valor introduït com una **cadena (`str`)**.

### Documentació Python de referència

- [`input()` — Built-in Functions](https://docs.python.org/3.14/library/functions.html#input)

````python
# Exemple
nom: str = input("Digues-me el teu nom:")
print(nom)
````

Per defecte, `input()` retorna sempre una cadena (`str`).

Si volem treballar amb un altre tipus de dada, hem de fer explícitament la conversió.

````python
# Exemple
edat: int = int(input("Digues-me la teva edat: "))
print(edat)
print(type(edat))
````

````python
# Exemple 2
descompte: float = float(input(
    "Digues-me quin descompte vols " \
    "per les rebaixes (per ex. 0.2 és un 20%) "))

print(descompte * 100, "%")
print(type(descompte))
````

````python
# Exemple 3: NO RECOMANABLE 
x = eval(input("Entra un valor"))  # NO USAR
#eval avalua l'entrada i fa que el tipus sigui l'adient
print(x)
type(x)

# x no es pot mostrar perquè eval() avalua una cadena d'entrada com a expressió de python
# llavors veu que l'input i l'interpreta com una variable, cosa que dona error perquè no la troba

# Atenció! eval pot executar el codi que entra l'usuari
# usar amb precaució!
````

### Introduir diverses dades

També podem demanar diverses dades en una sola entrada.

Una manera senzilla és demanar que l'usuari les separi amb un caràcter
determinat i després utilitzar el mètode `split()`.

Per exemple, podem demanar tres valors separats per comes:

```text
Anna,Cendra,Fuster
```

i separar-los amb:

```python
entrada.split(",")
```

Si volem eliminar possibles espais al principi o al final de cada valor,
podem aplicar també `strip()`.

Aquesta solució és preferible a utilitzar `eval()` sobre el text introduït
per l'usuari.

````python
# Exemple
entrada: str = input(
    "Digues-me el teu nom i cognoms separats per comes "
    "(per exemple, Anna,Cendra,Fuster): "
)

# es pot assignar múltiples variables amb una llista separada amb split() donat l'input
# per a cada valor recorregut, podem treure els espais en blanc inicials i finals prèviament amb strip()

nom, cognom1, cognom2 = [
    valor.strip()
    for valor in entrada.split(",")
]

print(nom)
print(cognom1)
print(cognom2)
````

## Output, sortida de dades: `print()`

La funció `print()` permet mostrar text i altres valors a la sortida.

Per exemple:

```python
print("Hola")
print(42)
print(variable)
```

Podem passar diversos valors separats per comes:

```python
print(nom, edat)
```

Per defecte, `print()`:

- separa els arguments amb un espai;
- afegeix un salt de línia al final.

També podem construir textos amb variables utilitzant **f-strings**:

```python
print(f"{nom} té {edat} anys")
```

### Documentació Python de referència

- [`print()` — Built-in Functions](https://docs.python.org/3.14/library/functions.html#print)
- [Formatted String Literals — f-strings](https://docs.python.org/3.14/tutorial/inputoutput.html#formatted-string-literals)
- [Input and Output](https://docs.python.org/3.14/tutorial/inputoutput.html)

````python
# Exemple
nom: str = "Jordi"
print(nom)
````

`print()` és una funció i, per tant, els seus arguments s'escriuen entre
parèntesis.

Els arguments poden ser, entre altres:

- literals:

```python
print("Hola")
print(17)
```

- variables:

```python
print(nom)
print(edat)
```

- expressions:

```python
print(17 - 2)
print(edat + 4)
```

`print()` converteix els objectes a string per mostrar-los a la sortida.

````python
# Exemple
edat: int = 14
nom: str = "Àlex"

print("Pere")
print(17)
print(nom)
print(edat)
print(17 - 2)
print(edat + 4)
print(nom + nom)
````

Per defecte, cada crida a `print()` acaba amb un salt de línia.

També podem mostrar diversos valors en una mateixa crida separant-los
amb comes:

```python
print(nom, edat)
```

Per defecte, `print()` insereix un espai entre els diferents arguments.

````python
# Exemple
edat: int = 14
nom: str = "alex"
print(nom, edat)
````

### F-strings i paràmetre `end`

Les **f-strings** permeten inserir el valor de variables o expressions directament dins d'una cadena.

Per exemple:

```python
print(f"{nom} té {edat} anys")
```

Les expressions que volem inserir s'escriuen entre claus `{}`.

El control del salt de línia, en canvi, **no depèn de les f-strings**.

`print()` disposa del paràmetre `end`, que indica què s'ha d'escriure després dels arguments.

Per defecte:

```python
end="\n"
```

és a dir, s'afegeix un salt de línia.

Si escrivim:

```python
print("Edat:", end=" ")
print(edat)
```

la segona crida a `print()` continua a la mateixa línia.

````python
# Exemple
edat: int = 18
nom: str = "Àlex"

print(f"{nom} té {edat} anys")
print(f"L'edat de {nom} és:", end=" ")
print(edat)
````

## ✍️ 1. Exercici 1: esport, joc o aplicació preferida

Fes una funció que demani a l'usuari informació sobre el seu esport,
joc o aplicació preferida i mostri les dades de manera clara a la pantalla.

Les dades han d'incloure, com a mínim:

- una cadena (`str`);
- un nombre (`int` o `float`);
- un valor booleà (`bool`).

Utilitza `input()` per obtenir les dades i una **f-string** per mostrar el resultat.

**Atenció!**: quan fem la conversió a booleà amb bool(input...) qualsevol cadena no buida es considerarà certa. Millor primer fer cadena: str = input(...) i després fer la conversió a booleà.

````python
def informacio_usuari() -> None:

    
    esport: str = input(f"Digues el teu esport preferit: ")
    temps: int = int(input(f"Quant de temps (en hores) li dediques a la setmana: "))
    pregunta: str = input(f"Voldries dedicar-ne més temps? (Respon Sí/No): ")
    resposta: bool = pregunta.capitalize() == "Sí"

    print()
    print("La teva informacio:")
    print(f"Esport preferit: {esport}.")
    print(f"Temps dedicat (en hores): {temps}.") 
    print(f"Més compromís: {resposta}.") 
    

informacio_usuari()
````

## ✍️ 2. Exercici 2: distància euclidiana

Escriu una funció `magnitude()` que accepti dos punts del pla i calculi
la distància euclidiana entre ells:

$$
d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}
$$

Cada punt es representarà amb una tupla de dos nombres:

```python
punt1 = (x1, y1)
punt2 = (x2, y2)
```

La funció ha de retornar la distància calculada com un `float`.

````python
def magnitude(
    punt1: tuple[float, float],
    punt2: tuple[float, float],
) -> float:
    """
    Calcula la distància euclidiana entre dos punts del pla.

    Parameters
    ----------
    punt1 : tuple[float, float]
        Coordenades (x, y) del primer punt.
    punt2 : tuple[float, float]
        Coordenades (x, y) del segon punt.

    Returns
    -------
    distancia : float
        Distància euclidiana entre els dos punts.
    """

    distancia_x = punt2[0]-punt1[0]
    distancia_y = punt2[1]-punt1[1]

    distancia_e = (distancia_x**2 + distancia_y**2) ** 0.5

    return distancia_e    
````
