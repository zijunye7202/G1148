Per comparar dos valors podem utilitzar els operadors de comparació:

- `<` menor que
- `<=` menor o igual que
- `==` igual que
- `>` major que
- `>=` major o igual que
- `!=` diferent de

Les comparacions retornen un valor booleà: **`True`** o **`False`**.

````python
# Exemple
print("4 == 2 + 2", 4 == 2 + 2)
print("4 != 2 + 2", 4 != 2 + 2)
print("4 < 2", 4 < 2)
print("4 <= 4", 4 <= 4)
print("4 > 2", 4 > 2)
print("4 >= 5", 4 >= 5)
````

Si el què comparem són cadenes de caràcters aquests operadors comparen l’ordre lexicogràfic dels caràcters que les formen. 

Per exemple “Hello” és més petit que “hello” i “a” és més petita que “b”.

Cal tenir en compte que majúscules i minúscules són caràcters diferents, i per tant, poden tenir un ordre lexicogràfic diferent.

````python
# Exemple
print("a < b", 'a'<'b')
print("hola == Hola", "hola"=="Hola")
print("hola > Hola", "hola">"Hola")

# Solem usar " per strings, i ' per caràcters

# L'ordre lexicogràfic depèn del valor UniCode d'un caràcter
# Com que 'A' = 65 i 'a' = 97, naturalment 'a' > 'A'
````

````python
print('A' < 'a')
````

També podem utilitzar els operadors booleans:

- **`and`**: combina dues condicions i és certa si totes dues són certes.
- **`or`**: combina dues condicions i és certa si almenys una és certa.
- **`not`**: nega una condició; transforma `True` en `False` i `False` en `True`.

Per exemple:

```python
a and b
a or b
not a

La precedència (jerarquia) dels operadors booleans és:

1.not
2.and
3.or

Per evitar ambigüitats, podem usar parèntesi: ```(a and b) or c```

````python
# Exemple
print("2 > 1 and 3 > 2", 2 > 1 and 3 > 2)
print("2 > 1 or 4 < 2", 2 > 1 or 4 < 2)
print("not 2 > 1", not 2 > 1)
````

Les taules de veritat d'aquests operadors són les següents:

Les taules de veritat dels operadors booleans són les següents.

### `not`

| P | not P |
|---|---|
| True | False |
| False | True |

### `and`

| P | Q | P and Q |
|---|---|---|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

### `or`

| P | Q | P or Q |
|---|---|---|
| True | True | True |
| True | False | True |
| False | True | True |
| False | False | False |

### XOR booleà (`!=`)

| P | Q | P != Q |
|---|---|---|
| True | True | False |
| True | False | True |
| False | True | True |
| False | False | False |





**Important:** `&`, `|` i `^` també poden aplicar-se a valors booleans, però són operadors **bit a bit**. Els veurem a continuació aplicats principalment a nombres enters.

**AVANÇAT:** Python també disposa d'operadors **bit a bit** (*bitwise*)
que treballen directament sobre la representació binària dels nombres enters:

- `&` AND bit a bit
- `|` OR bit a bit
- `^` XOR bit a bit
- `~` NOT bit a bit
- `<<` desplaçament a l'esquerra
- `>>` desplaçament a la dreta

Aquests operadors permeten crear **màscares de bits**.

Per exemple, podem utilitzar:

`n & 1`

per comprovar el darrer bit d'un nombre. Si el resultat és 1, el nombre
és senar; si és 0, és parell.

#### Convé diferenciar els operadors en tres famílies:
- Operadors de comparació: ==, !=, <, <=, >, >=. Comparen valors i produeixen un bool.
- Operadors lògics: and, or, not. Combinen o neguen condicions.
- Operadors bit a bit: &, |, ^, ~, <<, >>. Operen sobre els bits dels enters. No cal escriure en binari, Python els tracta internament de manera binària.

````python
# Exemple:
# L'expressió n & 1 retorna 1 si el nombre és senar
# i 0 si el nombre és parell.

# 00010100 &  (n = 20)
# 00000001    (1)
# --------
# 00000000    (= 0, 20 és parell)

# 00010101 &  (n = 21)
# 00000001    (1)
# --------
# 00000001    (= 1, 21 és senar)

print("20 & 1 =", 20 & 1)
print("21 & 1 =", 21 & 1)

# El nombre 1 actua com una màscara que conserva únicament
# el darrer bit del nombre.
````

````python
# Exemple:
# En ASCII, aplicar OR bit a bit amb el codi de l'espai
# a una lletra majúscula permet obtenir la minúscula corresponent.

# 01000001 |  ('A' = 65)
# 00100000    (' ' = 32)
# --------
# 01100001    ('a' = 97)

# La màscara 00100000 activa el bit que diferencia,
# en ASCII, una lletra majúscula de la seva minúscula.

# L'OR actua modificant el valor del binari si almeyns un dels nombres UniCode de la mateixa posició sigui un 1
# Només funciona amb els caràcters A-Z

for codi_caracter in range(65, 91):
    majuscula: str = chr(codi_caracter)
    minuscula: str = chr(codi_caracter | ord(" "))

    print(majuscula, minuscula)
````

## ✍️ 2. Exercici 2. Endevinar una paraula

Fes un programa per endevinar una paraula de quatre lletres.

Es demanarà a l'usuari que introdueixi una paraula i se li indicarà,
mitjançant valors booleans, quines lletres ha encertat.

Per cada lletra fallada perdrà una vida. Quan es quedi sense vides,
haurà de deixar de jugar.

Exemple:

```text
Endevina la paraula... HOLA

Has encertat dues lletres:
False False True True
Et queden 8 vides.

Endevina la paraula... MALA

Has encertat tres lletres:
False True True True
Et queden 7 vides.

````python
solucio: str = "CASA"
vides: int = 10

paraula: str = ""

# mentre que les vides siguin majors de 0 i no s'ha endevinat la paraula el joc segueix preguntant
# per mostrar els booleans i comparar cada lletre ho podem fer a través d'un for iteratiu
# posem la condició de mostrar un missatge si s'ha encertat la solució (totes les lletres coincideixen)
# llavors si alguna no coincideix es treu una vida, a més de mostrar els boleans

while solucio != paraula and vides > 0:
    paraula: str = (input("Endevina la paraula...")).upper()
    if solucio == paraula:
        print(f"Has encertat la solució {solucio}!")
    else:
        for lletra in range(0, len(paraula)):
            print(solucio[lletra] == paraula[lletra], end =" ")
            
        vides -= 1
        print()
        print(f"Et queden {vides} vides")

if vides == 0:
    print(f"T'has quedat sense vides, la solucio era {solucio}...")
````
