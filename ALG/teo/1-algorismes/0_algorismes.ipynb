## 🧠 Algorisme Rus per multiplicar dos enters

```python
def producte_rus(a: int, b: int) -> int:
    """ Multiplica dos enters a i b utilitzant l'algorisme del producte rus. 
    Retorna el resultat de a * b. """
    resultat = 0
    while a > 0:
        # Si a és senar, afegeix b al resultat
        if a % 2 == 1:
            resultat += b
        # Divideix a per 2 (divisió entera) i multiplica b per 2
        a //= 2
        b *= 2
    return resultat

producte_rus(3,1)
```
