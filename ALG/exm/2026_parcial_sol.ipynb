## Pregunta 2. Complexitat (1 punt)

Considera els dos fragments següents, on n és un enter positiu.

```python
# Algorisme A
for i in range(n):
    for j in range(n):
        print(i, j)
```

```python
# Algorisme B
for i in range(n):
    print(i)
for j in range(n):
    print(j)
```

a) Indica la complexitat de cadascun utilitzant la notació O gran. (0,5 punts)

b) Explica per què són diferents, malgrat que tots dos tenen dos bucles for. (0,5 punts)


Solució — Pregunta 2 (1 punt)

**a) Complexitat temporal (0,5 punts)**

- **Algorisme A:** $O(n^2)$, ja que el `print` s'executa $n \cdot n = n^2$ vegades. **0,25 punts.**
- **Algorisme B:** $O(n)$, ja que cada bucle executa $n$ iteracions i en total hi ha $2n$ execucions de `print`. **0,25 punts.**

**b) Justificació (0,5 punts)**

- **Algorisme A:** els bucles estan imbricats. Per cadascuna de les $n$ iteracions del bucle exterior s'executen $n$ iteracions del bucle interior. Per tant, els costos es multipliquen:

$$
T_A(n) = n \cdot n = n^2 \implies \boxed{O(n^2)}
$$

- **Algorisme B:** els bucles són consecutius. Els costos se sumen i el factor constant no canvia l'ordre de complexitat:

$$
T_B(n) = n+n = 2n \implies \boxed{O(n)}
$$


## Pregunta 3. Algorismes numèrics (1 punt)

Volem determinar si un nombre enter positiu N és primer.

+ a) Explica per què, en un algorisme que comprova possibles divisors, és suficient provar divisors fins a $sqrt(n)$, en lloc de provar-los tots fins a N-1. (0,6 punts)

+ b) Si utilitzem aquest procediment, quina és la seva complexitat temporal en funció del valor de N, suposant que cada comprovació de divisibilitat té cost constant? (0,4 punts)


### Solució — Pregunta 3 (1 punt)

**a) Justificació del límit $\sqrt{N}$ (0,6 punts)**

Si $N$ és compost, existeixen enters $a,b>1$ tals que:

$$
N = a \cdot b
$$

No poden ser tots dos estrictament més grans que $\sqrt{N}$, perquè aleshores:

$$
a > \sqrt{N} \quad \text{i} \quad b > \sqrt{N}
\implies a \cdot b > N
$$

Per tant, almenys un divisor és menor o igual que $\sqrt{N}$.

Així, si no trobem cap divisor entre $2$ i $\lfloor\sqrt{N}\rfloor$, podem concloure que $N$ és primer (per a $N \geq 2$).

**b) Complexitat temporal (0,4 punts)**

En el pitjor cas es proven de l'ordre de $\sqrt{N}$ possibles divisors. Si cada comprovació de divisibilitat té cost constant, la complexitat temporal és:

$$
\boxed{O(\sqrt{N})}
$$

Aquesta complexitat es mesura en funció del **valor numèric de $N$**, no del nombre de bits necessaris per representar-lo.

## Exercici 1. Validació d'un codi d'identificació (5,5 punts)

Una empresa assigna als seus productes un codi d'identificació format per una seqüència de xifres i un dígit de control final. Aquest dígit permet detectar alguns errors en la introducció dels codis.

El dígit de control es calcula mitjançant el procediment següent:
1. Es recorren les xifres del codi, excloent-ne el dígit de control, d'esquerra a dreta.
2. Cada xifra es multiplica per la seva posició, començant per 1.
3. Se sumen tots els productes obtinguts.
4. El dígit de control és el residu de dividir aquesta suma entre 10.


Exemple

Per al codi 5274, la suma ponderada és:
$$S=5\cdot1+2\cdot2+7\cdot3+4\cdot4=46$$

El dígit de control és:
$$d=46\bmod10=6$$

Per tant, el codi complet vàlid és 52746.

Es demana que implementis una funció que rep un codi complet, amb el dígit de control a l'última posició, i retorna True si el dígit de control és correcte i False en cas contrari.

Condicions:
- El codi conté exclusivament xifres del 0 al 9.
- El codi té almenys dos xifres.
- El codi pot començar per zeros.
- No es poden utilitzar biblioteques externes.

Fes servir la plantilla següent:

```python
def valida_codi(codi: str) -> bool:
    """
    Comprova la validesa del dígit de control d'un codi.

    El dígit de control és el residu mòdul 10 de la suma
    de les xifres del codi, ponderades per la seva posició.

    Es pressuposa que el codi conté almenys dues xifres
    i que tots els seus caràcters són dígits decimals.

    :param codi: Cadena de xifres amb el dígit de control final.
    :return: True si el dígit de control és correcte;
        False en cas contrari.
    """
```

```python
# Solució Exercici 1

def valida_codi(codi: str) -> bool:
    """
    Comprova la validesa del dígit de control d'un codi.

    El dígit de control és el residu mòdul 10 de la suma
    de les xifres del codi, ponderades per la seva posició.

    Es pressuposa que el codi conté almenys dues xifres
    i que tots els seus caràcters són dígits decimals.

    :param codi: Cadena de xifres amb el dígit de control final.
    :return: True si el dígit de control és correcte;
        False en cas contrari.
    """
    suma: int = 0

    # Recorrem totes les xifres excepte la darrera.
    for i in range(len(codi) - 1):
        xifra: int = int(codi[i])
        suma += xifra * (i + 1)

    # Calculem el dígit de control esperat.
    control: int = suma % 10

    # Comparem amb el darrer dígit del codi.
    return control == int(codi[-1])
```

```python
# Asserts públics

assert valida_codi("52746") is True
assert valida_codi("52743") is False
assert valida_codi("1234") is True
assert valida_codi("1235") is False
assert valida_codi("00") is True
assert valida_codi("99") is True
```

## Exercici 3. Aplicació de seguretat (3,5 punts)

Una aplicació de seguretat analitza el contingut textual de fitxers per detectar signatures que identifiquen possibles amenaces.

La funció ha de cercar la signatura dins del text amb un algorisme més eficient que la cerca ingènua. Recordeu els algorismes treballats a classe de cerca d'un patró exacte i mireu la xuleta de l'examen per als algorismes base ja resolts.

La funció ha de retornar la posició inicial de la signatura dins del text, començant a comptar des de zero. Si no es troba, ha de retornar -1.

Podem suposar que:
- Tant el text com la signatura són cadenes no buides.
- Les cadenes contenen únicament caràcters ASCII.
- Es garanteix que la signatura apareix, com a màxim, una vegada al text, sense distingir majúscules i minúscules.
- No es permet utilitzar funcions de Python que resolen el quid de l'exercici.

Fes servir la plantilla següent:

```python
def localitza_signatura(text: str, signatura: str) -> int:
    """
    Localitza una signatura en un text.

    Es pressuposa que les dues cadenes són no buides,
    contenen només caràcters ASCII i que la signatura
    apareix com a màxim una vegada.

    :param text: Cadena en què es busca la signatura.
    :param signatura: Cadena que es vol localitzar.
    :return: Posició inicial de la signatura dins del text,
        o -1 si no apareix.
    """
```

```python
# Solució exercici 3

def localitza_signatura(text: str, signatura: str) -> int:
    """
    Localitza una signatura en un text.

    Es pressuposa que les dues cadenes són no buides,
    contenen només caràcters ASCII i que la signatura
    apareix com a màxim una vegada.

    :param text: Cadena en què es busca la signatura.
    :param signatura: Cadena que es vol localitzar.
    :return: Posició inicial de la signatura dins del text,
        o -1 si no apareix.
    """
    # Normalitzem les cadenes per ignorar majúscules.
    text = text.upper()
    signatura = signatura.upper()

    m: int = len(signatura)
    n: int = len(text)

    if m > n:
        return -1

    # Construïm la taula de desplaçaments ASCII.
    taula_desplacaments: list[int] = []

    for k in range(256):
        taula_desplacaments.append(m)

    # Calculem els salts dels caràcters de la signatura.
    for k in range(m - 1):
        taula_desplacaments[ord(signatura[k])] = m - k - 1

    taula_desplacaments = tuple(taula_desplacaments)

    # Posició del text alineada amb el final del patró.
    k: int = m - 1

    while k < n:
        j: int = m - 1
        i: int = k

        # Comparem de dreta a esquerra.
        while j >= 0 and text[i] == signatura[j]:
            j -= 1
            i -= 1

        # Hem trobat la signatura completa.
        if j == -1:
            return i + 1

        # Apliquem el desplaçament de Horspool.
        k += taula_desplacaments[ord(text[k])]

    return -1
```

```python
# Asserts privats

# Casos de longitud mínima.
assert localitza_signatura("A", "a") == 0
assert localitza_signatura("a", "B") == -1
assert localitza_signatura("abc", "ABCD") == -1

# Signatura al principi, al mig i al final.
assert localitza_signatura("AbCdZZZZ", "abcd") == 0
assert localitza_signatura("ZZabCDzz", "ABcd") == 2
assert localitza_signatura("zzzzABcd", "abcd") == 4

# Signatura igual a tot el text.
assert localitza_signatura("aBcDeF", "AbCdEf") == 0

# Repeticions internes en la signatura.
assert localitza_signatura("xxabABczz", "ABABC") == 2
assert localitza_signatura("zzzzAaab", "aaab") == 4
assert localitza_signatura("AAAAAAAA", "aaab") == -1

# Coincidències parcials abans de la correcta.
assert localitza_signatura("ABCXABCYABCD", "aBcD") == 8
assert localitza_signatura("ABABABAC", "abac") == 4
assert localitza_signatura("ABABABAX", "abac") == -1

# Diferents combinacions de majúscules i minúscules.
assert localitza_signatura("abcdefGHI", "DeFg") == 3
assert localitza_signatura("ABCdef", "DEF") == 3
assert localitza_signatura("abcdefgh", "XYZ") == -1

# Espais i signes de puntuació.
assert localitza_signatura("abc!def?ghi", "GH") == 8

# Signatura d'un únic caràcter.
assert localitza_signatura("abcdefgh", "H") == 7
assert localitza_signatura("ABCDEFGH", "z") == -1
```

Calcula i raona la complexitat del teu codi amb O gran. Explica què és la n o altres variables que apareguin a la complexitat.

Solució exercici 4. Cerca de signatures amb Horspool

Paràmetres:
- \(n\): longitud del text.
- \(m\): longitud de la signatura.
Càlcul dels passos:
1. Convertir el text i la signatura a majúscules: \(n+m\) operacions.

3. Construir la taula de desplaçaments: \(m-1\) operacions.

\[
O(n+m+256+m+nm)=\boxed{O(nm)}
\]

El cost dominant és la cerca. Encara que Horspool acostuma a evitar moltes comparacions gràcies als salts, en el pitjor cas la complexitat és \(O(nm)\).

## Exercici 5. Problema repte: Primers bessons (+1 punt de la part pràctica)

Dos nombres primers són bessons si la seva diferència és exactament 2. Per exemple, 11 i 13 són primers bessons.

Implementa una funció que retorna una llista ordenada de totes les parelles de primers bessons en què tots dos nombres són estrictament menors que n, amb n >= 2.

Pots usar la funció de Python sorted(col.lecció) que retorna la col.lecció ordenada de menor a major.

Fes servir la plantilla següent:

```python
# Exemple de l'ús de sorted

Conjunt_A = (1, 6, 7, 9, 3, 2)
Conjunt_A_ordenat = sorted(Conjunt_A)
print(Conjunt_A_ordenat)
```

```python
def primers_bessons(n: int) -> list[list[int, int]]:
    """
    Calcula les parelles de primers bessons menors que n.

    Es pressuposa que n és més gran o igual que 2.

    :param n: Límit superior exclusiu dels nombres primers.
    :return: Llista ordenada de parelles de primers bessons.
    """
```

```python
# Solució exercici 5

def era(n: int) -> set[int]:
    """
    Calcula els nombres primers estrictament menors que n.

    :param n: Límit superior exclusiu.
    :return: Conjunt dels nombres primers menors que n.
    """
    era1: set[int] = set(range(2, n))
    era2: set[int] = {
        i
        for q in range(2, int(n ** 0.5) + 1)
        for i in range(q * 2, n, q)
    }

    return era1 - era2


def primers_bessons(n: int) -> list[list[int, int]]:
    """
    Calcula les parelles de primers bessons menors que n.

    Es pressuposa que n és més gran o igual que 2.

    :param n: Límit superior exclusiu dels nombres primers.
    :return: Llista ordenada de parelles de primers bessons.
    """
    primers: set[int] = era(n)
    bessons: list[list[int, int]] = []

    # Comprovem si cada primer té un primer bessó superior.
    for p in sorted(primers):
        if p + 2 in primers:
            bessons.append([p, p + 2])

    return bessons
```

```python
# Asserts públics

assert primers_bessons(2) == []
assert primers_bessons(5) == []
assert primers_bessons(6) == [[3, 5]]
assert primers_bessons(10) == [[3, 5], [5, 7]]
assert primers_bessons(20) == [
    [3, 5], [5, 7], [11, 13], [17, 19]
]
assert primers_bessons(30) == [
    [3, 5], [5, 7], [11, 13], [17, 19]
]
```

```python
# Límits inferiors.
assert primers_bessons(3) == []
assert primers_bessons(4) == []
assert primers_bessons(7) == [[3, 5]]
assert primers_bessons(8) == [[3, 5], [5, 7]]

# Límit superior exclusiu.
assert primers_bessons(13) == [[3, 5], [5, 7]]
assert primers_bessons(14) == [
    [3, 5], [5, 7], [11, 13]
]
assert primers_bessons(19) == [
    [3, 5], [5, 7], [11, 13]
]

# Intervals més llargs i ordenació.
assert primers_bessons(50) == [
    [3, 5], [5, 7], [11, 13],
    [17, 19], [29, 31], [41, 43]
]

assert primers_bessons(100) == [
    [3, 5], [5, 7], [11, 13],
    [17, 19], [29, 31], [41, 43],
    [59, 61], [71, 73]
]

# El límit 101 no incorpora noves parelles.
assert primers_bessons(101) == primers_bessons(100)

# Parelles noves més enllà de 100.
assert primers_bessons(110) == [
    [3, 5], [5, 7], [11, 13],
    [17, 19], [29, 31], [41, 43],
    [59, 61], [71, 73], [101, 103],
    [107, 109]
]

# Comprovació de l'última parella sota 200.
assert primers_bessons(200)[-1] == [197, 199]

# Comprovació del nombre total de parelles.
assert len(primers_bessons(200)) == 15
```
