## ✍️ Factors d'un nombre

Donat un natural major que 1, descompon-lo progressivament en factors i retorna tots els seus factors primers.


### 🧠 Tasca


```python
def factors_primers(n) -> list[int]:
    """
    Aquesta funció 
    calcula els factors primers 
    d'un nombre natural donat.
    
    Parameters
    ----------
    n: int
    
    Returns
    -------
    factors: list[int]
    """

    # n = 60
    # sabem que n = xq + r
    # anem dividint n en factors primers començant amb divisor x = 2
    # en cas de que no es pogui dividir entre x aleshores augmentem el divisor += 1
    # els factors primers seran els divisors que vaig afegint 
    
    # tenint en compte que si n es compost llavors n = a * b
    # llavors com a mínim un dels valors a, b <= (n ** 0.5)
    # en comptes de recórrer tots els factors primers de 2 a N
    # ho fem de 2 fins a (n) ** 0.5
    # llavors per a aquells valors que superi l'arrel de N implica que sigui primer

    divisor: int = 2
    factors: list[int] = []

    while divisor <= (n ** 0.5):

        if n % divisor == 0:
            factors += [divisor]
            n //= divisor

        else:
            divisor += 1

    # la n resultant surt fora del bucle després de dividir iterativament pels factors
    # si el valor d'n és major a l'arrel de n inicial, és necessari que sigui primer
    
    if n > 1:

        factors += [n]

    return factors 

factors_primers(12)
```

### ✅ Asserts públics


```python
def factors_primers(n: int) -> list[int]:

    # funcio que retorna factors primers d'n
    # donat que n = xq + r
    # iniciem el divisor x = 2 com a primer factor
    # si es pot dividir entre x llavors afegim dins dels factors
    # dividim n // x
    # en cas contrari augmentem el valor de x
    
    # si n és un nombre compost aleshores n = a * b
    # com a mínim un dels valors a, b <= n ^ 0.5
    # és una condició per tal de procedir amb les division iteratives
    # això evita iterar factors de 2 fins N

    # n surt del bucle quan sigui major a l'arrel de n inicial
    # i és necessari que sigui un factor també

    factors: list[int] = []
    x: int = 2

    while x <= (n ** 0.5):

        if n % x == 0: 

            factors += [x] 
            n //= x

        else:
            x += 1 

    if n > 1: 

        factors += [n] 

    return factors 
```

```python
assert factors_primers(600851475143) == [71, 839, 1471, 6857]
assert factors_primers(12) == [2, 2, 3]
```

### 📊 Complexitat

Indica i justifica la complexitat temporal i espacial de la solució.


```python
# COMPLEXITAT TEMPORAL
# Les operacions simples tenen un cost O(1).
# El bucle while comprova divisors fins a n ** 0.5.
# En el pitjor cas, n és primer i no trobem divisors.
# Les divisions que redueixen n aporten O(log n).
# T(n) = O(sqrt(n) + log(n)) = O(sqrt(n))

# COMPLEXITAT ESPACIAL
# Les variables n i x ocupen O(1) espai
# en el model simplificat.
# La llista factors acumula com a màxim log2(n)
# factors primers, comptant repeticions.
# S(n) = O(1) + O(log n) = O(log n).
```
