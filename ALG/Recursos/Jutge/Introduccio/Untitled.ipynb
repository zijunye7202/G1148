```python
a: int = int(input())

print("YES") if ((a % 400 == 0) or (a % 4 == 0 and a % 100 != 0)) else print("NO")
```

```python
a1, a2, b1, b2 = map(int, input().split())

if (a1 <= b2 and a2 >= b1):

    print(f"[{max(a1, b1)},{min(a2, b2)}]")

else:
    print("[]")
```

```python
a1, a2, b1, b2 = map(int, input().split())

if (a1 > b1 and a2 < b2):
    print("1")

elif (b1 > a1 and b2 < a2):
    print("2")

elif (a1 == b1 and a2 == b2):
    print("=")

else: 
    print("?") 
```

```python
n: int = int(input())
for i in range (n + 1):
    print(i)
```

```python
x, y = map(int, input().split())

if x > y:
    for i in range (x, y-1, -1):
        print(i)
else:
    for i in range (y, x-1, -1):
        print(i)
```

```python
a, b = map(int, input().split())

if a <= b:
    for i in range(a, b+1):
        if i == b:
            print(i)
        else:
            print(i, end=",")
else:
    print("")
```
